## 📊 SQL Analysis — Queries to Execute

1. Show the total number of flights for each aircraft model, listing the model and its count.
2. List all aircraft (registration, model) that have been assigned to more than 5 flights.
3. For each airport, display its name and the number of outbound flights, but only for airports with more than 5 flights.
4. Find the top 3 destination airports (name, city) by number of arriving flights, sorted by count descending.
5. Show for each flight: number, origin, destination, and a label 'Domestic' or 'International' using `CASE WHEN` on country match.
6. Show the 5 most recent arrivals at "DEL" airport including flight number, aircraft, departure airport name, and arrival time, ordered by latest arrival.
7. Find all airports with no arriving flights (never used as a destination in the flights table).
8. For each airline, count the number of flights by status (e.g., 'On Time', 'Delayed', 'Cancelled') using `CASE WHEN`.
9. Show all cancelled flights, with aircraft and both airports, ordered by departure time descending.
10. List all city pairs (origin–destination) that have more than 2 different aircraft models operating flights between them.
11. For each destination airport, compute the % of delayed flights (status='Delayed') among all arrivals, sorted by highest percentage.

## 0. Configuration & Helpers

Shared settings, the airport list and the `run_query()` / `api_get()` helpers used by every section.

````python
import os
import time
import math
import requests
import pandas as pd
import mysql.connector
from datetime import datetime, timedelta, timezone
from decimal import Decimal
from zoneinfo import ZoneInfo

# ---------------------------------------------------------------
# Database settings. The password is read from an environment
# variable (set once with:  setx DB_PASSWORD "your_password").
# ---------------------------------------------------------------
DB_NAME = "flight_analytics"
DB_CONFIG = {
    "host": "localhost",
    "user": "root",
    "password": os.getenv("DB_PASSWORD"),
}

# ---------------------------------------------------------------
# AeroDataBox (RapidAPI) settings. Calls rotate across the keys,
# which spreads the per-second rate limit and the monthly quota.
# ---------------------------------------------------------------
API_HOST = "aerodatabox.p.rapidapi.com"
BASE_URL = f"https://{API_HOST}"
API_KEYS = [k.strip() for k in os.getenv("RAPIDAPI_KEYS", "").split(",") if k.strip()] or [
    "b15c300a6bmsh5b86ed4f7051c47p1b9521jsn506aa8bcd035",
    "232e995e0cmshd34dc590f07dfedp1cf8f5jsne2fe2f5f3d98",
]
CALL_DELAY_SEC = 1.2        # pause between API calls (per-second rate limit)
MIN_UNITS_RESERVE = 30      # stop making optional calls below this many units

# ---------------------------------------------------------------
# Project settings: ONE list of airports used by every section.
# DEL (VIDP) is included because SQL question 6 asks about it.
# ---------------------------------------------------------------
AIRPORT_ICAO_CODES = [
    "BGAA",  # JEG  Aasiaat
    "VIDP",  # DEL  New Delhi
    "EGLC",  # LCY  London City
    "EGLL",  # LHR  London Heathrow
    "DTNH",  # NBE  Enfidha
    "EBAW",  # ANR  Antwerp
    "FACT",  # CPT  Cape Town
    "GCRR",  # ACE  Lanzarote
    "OMDB",  # DXB  Dubai
    "KJFK",  # JFK  New York
    "KLYH",  # LYH  Lynchburg
    "LCLK",  # LCA  Larnaca
    "MMBT",  # HUX  Huatulco
    "OEJN",  # JED  Jeddah
    "PAJN",  # JNU  Juneau
]
HOURS_PER_WINDOW = 12       # the API allows at most 12 hours per flights call
WINDOWS_PER_AIRPORT = 4     # 4 x 12 h = the past 2 days of flights
DELAY_THRESHOLD_MIN = 15    # a flight counts as delayed above this many minutes
MAX_AIRCRAFT_CALLS = 150    # cap on aircraft look-ups per run (API quota)

_key_index = 0
UNITS_LEFT = {}


def get_connection(with_database=True):
    """Open a MySQL connection (database optional so it can be created)."""
    config = dict(DB_CONFIG)
    if with_database:
        config["database"] = DB_NAME
    return mysql.connector.connect(**config)


def run_query(sql, params=None):
    """Run a SELECT and return the result as a DataFrame."""
    conn = get_connection()
    try:
        cur = conn.cursor()
        cur.execute(sql, params or ())
        columns = [d[0] for d in cur.description]
        df = pd.DataFrame(cur.fetchall(), columns=columns)
    finally:
        conn.close()
    for col in df.columns:
        if df[col].map(lambda v: isinstance(v, Decimal)).any():
            df[col] = pd.to_numeric(df[col].astype(float), errors="coerce")
    return df


def api_get(path, params=None, max_retries=6):
    """GET an AeroDataBox endpoint.

    Retries on rate limits (429) and server errors, rotating API keys.
    Returns parsed JSON, or None when the endpoint has no data (400/404)
    or the response is not JSON.
    """
    global _key_index
    last_error = None
    for attempt in range(max_retries):
        key = API_KEYS[_key_index % len(API_KEYS)]
        _key_index += 1
        time.sleep(CALL_DELAY_SEC)
        try:
            response = requests.get(
                BASE_URL + path,
                headers={"x-rapidapi-key": key, "x-rapidapi-host": API_HOST},
                params=params,
                timeout=40,
            )
        except requests.RequestException as err:
            last_error = err
            time.sleep(2 ** attempt)
            continue

        left = response.headers.get("X-RateLimit-API-Units-Remaining")
        if left is not None:
            UNITS_LEFT[key[-6:]] = int(left)

        if response.status_code == 200:
            try:
                return response.json()
            except ValueError:
                return None
        if response.status_code in (204, 400, 404):      # 204 = no flights in that window
            print(f"   no data ({response.status_code}) for {path}")
            return None
        last_error = f"HTTP {response.status_code}"
        time.sleep(2 ** attempt)          # 429 / 5xx: back off and retry
    raise RuntimeError(f"API call failed after {max_retries} tries: {path} ({last_error})")


def units_remaining():
    """Smallest remaining API-unit count seen across the keys."""
    return min(UNITS_LEFT.values()) if UNITS_LEFT else None


print("Configuration loaded.")
print("Airports configured :", len(AIRPORT_ICAO_CODES))
print("API keys configured :", len(API_KEYS))
print("DB password found   :", DB_CONFIG["password"] is not None)
````

## 1. SQL Table Setup

````python
RESET_TABLES = False   # True drops airport, flights and airport_delays and rebuilds them

conn = get_connection(with_database=False)
cursor = conn.cursor()

cursor.execute(f"CREATE DATABASE IF NOT EXISTS {DB_NAME}")
cursor.execute(f"USE {DB_NAME}")
print(f"Database '{DB_NAME}' ready.")

if RESET_TABLES:
    for table in ("flights", "airport_delays", "airport"):
        cursor.execute(f"DROP TABLE IF EXISTS {table}")
    print("Tables dropped (RESET_TABLES = True).")

cursor.execute("""
CREATE TABLE IF NOT EXISTS airport (
    airport_id   INT AUTO_INCREMENT PRIMARY KEY,
    icao_code    VARCHAR(10) UNIQUE,
    iata_code    VARCHAR(10) UNIQUE,
    name         VARCHAR(255),
    city         VARCHAR(255),
    country      VARCHAR(255),
    country_code VARCHAR(5),
    continent    VARCHAR(50),
    latitude     DOUBLE,
    longitude    DOUBLE,
    timezone     VARCHAR(50)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS aircraft (
    aircraft_id    INT AUTO_INCREMENT PRIMARY KEY,
    registration   VARCHAR(30) UNIQUE,
    model          VARCHAR(255),
    manufacturer   VARCHAR(255),
    icao_type_code VARCHAR(255),
    owner          VARCHAR(255)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS flights (
    flight_id             VARCHAR(100) PRIMARY KEY,
    flight_number         VARCHAR(20),
    aircraft_registration VARCHAR(30),
    origin_iata           VARCHAR(10),
    destination_iata      VARCHAR(10),
    scheduled_departure   DATETIME,
    actual_departure      DATETIME,
    scheduled_arrival     DATETIME,
    actual_arrival        DATETIME,
    status                VARCHAR(30),
    airline_code          VARCHAR(10),
    aircraft_model        VARCHAR(100),
    origin_airport_name   VARCHAR(150),
    destination_airport_name VARCHAR(150),
    origin_country_code   VARCHAR(5),
    destination_country_code VARCHAR(5),
    delay_minutes         INT,
    INDEX idx_origin (origin_iata),
    INDEX idx_destination (destination_iata),
    INDEX idx_status (status),
    INDEX idx_airline (airline_code),
    INDEX idx_registration (aircraft_registration)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS airport_delays (
    delay_id         INT AUTO_INCREMENT PRIMARY KEY,
    airport_iata     VARCHAR(10),
    delay_date       DATE,
    source           VARCHAR(20),
    total_flights    INT,
    delayed_flights  INT,
    avg_delay_min    DECIMAL(8,2),
    median_delay_min DECIMAL(8,2),
    canceled_flights INT,
    UNIQUE KEY uq_airport_day (airport_iata, delay_date, source)
)
""")

conn.commit()

cursor.execute("SHOW TABLES")
print("Tables:", [t[0] for t in cursor.fetchall()])

cursor.close()
conn.close()
print("Database setup completed successfully.")
````

## 2. Airport Data Extraction

````python
def get_airport_data(icao_code):
    return api_get(f"/airports/icao/{icao_code}")


def store_airport(cursor, data):
    sql = """
    INSERT INTO airport
        (icao_code, iata_code, name, city, country, country_code,
         continent, latitude, longitude, timezone)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
    ON DUPLICATE KEY UPDATE
        iata_code    = VALUES(iata_code),
        name         = VALUES(name),
        city         = VALUES(city),
        country      = VALUES(country),
        country_code = VALUES(country_code),
        continent    = VALUES(continent),
        latitude     = VALUES(latitude),
        longitude    = VALUES(longitude),
        timezone     = VALUES(timezone)
    """
    location = data.get("location") or {}
    country = data.get("country") or {}
    values = (
        data.get("icao"),
        data.get("iata"),
        data.get("fullName"),
        data.get("municipalityName"),
        country.get("name"),
        country.get("code"),
        (data.get("continent") or {}).get("name"),
        location.get("lat"),
        location.get("lon"),
        data.get("timeZone"),
    )
    cursor.execute(sql, values)


conn = get_connection()
cursor = conn.cursor()

success, failed = [], []

for icao in AIRPORT_ICAO_CODES:
    try:
        airport = get_airport_data(icao)
        if not airport:
            raise ValueError("empty response")
        store_airport(cursor, airport)
        conn.commit()
        print(f"Saved  : {airport.get('fullName')} ({icao} / {airport.get('iata')})")
        success.append(icao)
    except Exception as err:
        print(f"Failed : {icao} -> {err}")
        failed.append(icao)

print("\n" + "=" * 50)
print("VERIFYING DATA IN DATABASE")
print("=" * 50)
cursor.close()
conn.close()

print(run_query("""
    SELECT airport_id, icao_code, iata_code, name, city, country, timezone
    FROM airport ORDER BY airport_id
""").to_string(index=False))

print(f"\nTotal airports : {len(AIRPORT_ICAO_CODES)}")
print(f"Successful     : {len(success)}")
print(f"Failed         : {len(failed)} {failed if failed else ''}")
print(f"API units left : {units_remaining()}")
````

## 3. Flight Data Extraction

The API's `movement.airport` is the *other* airport on the trip (the destination for a departure, the origin for an arrival), so the queried airport is placed at the opposite end.

````python
airports_df = run_query("""
    SELECT iata_code, icao_code, name, country_code, timezone
    FROM airport WHERE iata_code IS NOT NULL
""")
airports_df = airports_df[airports_df["icao_code"].isin(AIRPORT_ICAO_CODES)]
print("Airports to fetch flights for:", len(airports_df))


def parse_time(block):
    """Convert an API time block {'utc': '2026-11-01 17:30Z'} to a naive UTC datetime."""
    if not block:
        return None
    ts = pd.to_datetime(block.get("utc"), utc=True, errors="coerce")
    if pd.isna(ts):
        return None
    return ts.tz_localize(None).to_pydatetime()


def movement_to_row(item, airport, direction):
    """Turn one API flight record into one flat row.

    NOTE: in the API, movement.airport is the OTHER airport on the trip:
    the destination for a departure, the origin for an arrival.
    The airport we queried is the opposite end.
    """
    movement = item.get("movement") or {}
    other = movement.get("airport") or {}
    other_code = other.get("iata") or other.get("icao")
    number = item.get("number")
    if not other_code or not number:
        return None

    scheduled = parse_time(movement.get("scheduledTime"))
    actual = parse_time(movement.get("runwayTime") or movement.get("revisedTime"))
    status = item.get("status")
    if status and status.startswith("Cancel"):
        actual = None
    delay = None
    if scheduled and actual:
        delay = int((actual - scheduled).total_seconds() // 60)

    aircraft = item.get("aircraft") or {}
    other_country = (other.get("countryCode") or "").upper() or None
    home_country = airport["country_code"]

    row = {
        "flight_number": number,
        "aircraft_registration": aircraft.get("reg"),
        "aircraft_model": aircraft.get("model"),
        "airline_code": (item.get("airline") or {}).get("iata"),
        "status": status,
        "delay_minutes": delay,
        "scheduled_departure": None, "actual_departure": None,
        "scheduled_arrival": None, "actual_arrival": None,
    }
    if direction == "Departure":
        row.update(
            origin_iata=airport["iata_code"], destination_iata=other_code,
            origin_airport_name=airport["name"], destination_airport_name=other.get("name"),
            origin_country_code=home_country, destination_country_code=other_country,
            scheduled_departure=scheduled, actual_departure=actual,
        )
    else:
        row.update(
            origin_iata=other_code, destination_iata=airport["iata_code"],
            origin_airport_name=other.get("name"), destination_airport_name=airport["name"],
            origin_country_code=other_country, destination_country_code=home_country,
            scheduled_arrival=scheduled, actual_arrival=actual,
        )
    day = scheduled.strftime("%Y%m%d") if scheduled else "NA"
    row["flight_id"] = f"{number.replace(' ', '')}_{row['origin_iata']}_{row['destination_iata']}_{day}"
    return row


def local_windows(timezone_name):
    """Yield (from, to) local-time strings for the past WINDOWS_PER_AIRPORT windows."""
    try:
        tz = ZoneInfo(timezone_name)
    except Exception:
        tz = timezone.utc
    end = datetime.now(tz).replace(second=0, microsecond=0)
    for i in range(WINDOWS_PER_AIRPORT):
        to_time = end - timedelta(hours=HOURS_PER_WINDOW * i)
        from_time = to_time - timedelta(hours=HOURS_PER_WINDOW) + timedelta(minutes=1)
        yield from_time.strftime("%Y-%m-%dT%H:%M"), to_time.strftime("%Y-%m-%dT%H:%M")


rows, successful, failed = [], [], []

for _, airport in airports_df.iterrows():
    iata = airport["iata_code"]
    got_any = False
    for start, end in local_windows(airport["timezone"]):
        try:
            data = api_get(
                f"/flights/airports/iata/{iata}/{start}/{end}",
                params={"direction": "Both", "withCancelled": "true",
                        "withCodeshared": "false", "withCargo": "false",
                        "withPrivate": "false", "withLocation": "false"},
            )
        except Exception as err:
            print(f"Failed : {iata} {start} -> {err}")
            failed.append(f"{iata} {start}")
            continue
        if not data:
            continue
        got_any = True
        for item in data.get("departures", []):
            row = movement_to_row(item, airport, "Departure")
            if row:
                rows.append(row)
        for item in data.get("arrivals", []):
            row = movement_to_row(item, airport, "Arrival")
            if row:
                rows.append(row)
    print(f"Fetched {iata}: total rows so far = {len(rows)}  (API units left: {units_remaining()})")
    if got_any:
        successful.append(iata)

columns = [
    "flight_id", "flight_number", "aircraft_registration", "origin_iata", "destination_iata",
    "scheduled_departure", "actual_departure", "scheduled_arrival", "actual_arrival",
    "status", "airline_code", "aircraft_model", "origin_airport_name", "destination_airport_name",
    "origin_country_code", "destination_country_code", "delay_minutes",
]

df = pd.DataFrame(rows)
print(f"\nRaw rows collected : {len(df)}")

if df.empty:
    print("No flights were collected - check the API key / quota and re-run this cell.")
else:
    # A flight between two of our airports is seen twice (once as a departure,
    # once as an arrival). Merge those into a single row, keeping every known time.
    df = df.groupby("flight_id", as_index=False).first()
    df = df[columns].astype(object).where(df[columns].notna(), None)
    print(f"Unique flights     : {len(df)}")

    insert_sql = f"""
    INSERT INTO flights ({", ".join(columns)})
    VALUES ({", ".join(["%s"] * len(columns))})
    ON DUPLICATE KEY UPDATE
        {", ".join(f"{c} = VALUES({c})" for c in columns if c != "flight_id")}
    """
    conn = get_connection()
    cursor = conn.cursor()
    records = list(df.itertuples(index=False, name=None))
    for i in range(0, len(records), 1000):
        cursor.executemany(insert_sql, records[i:i + 1000])
    conn.commit()
    cursor.close()
    conn.close()

    total = run_query("SELECT COUNT(*) AS n FROM flights")["n"][0]
    print(f"Rows present in flights table : {total}")

print(f"\nSuccessful airports : {len(successful)}")
print(f"Failed windows      : {len(failed)} {failed if failed else ''}")
print(f"API units left      : {units_remaining()}")
````

## 4. Aircraft Data Extraction

````python
def manufacturer_from(type_name):
    """The API has no manufacturer field, so take it from the type name ('Airbus A320' -> 'Airbus')."""
    if not type_name:
        return None
    for maker in ("De Havilland", "Mitsubishi", "Antonov"):
        if type_name.startswith(maker):
            return maker
    return type_name.split()[0]


def get_aircraft_data(registration):
    return api_get(f"/aircrafts/reg/{registration}")


def store_aircraft(cursor, data):
    sql = """
    INSERT INTO aircraft (registration, model, manufacturer, icao_type_code, owner)
    VALUES (%s, %s, %s, %s, %s)
    ON DUPLICATE KEY UPDATE
        model          = VALUES(model),
        manufacturer   = VALUES(manufacturer),
        icao_type_code = VALUES(icao_type_code),
        owner          = VALUES(owner)
    """
    type_name = data.get("typeName") or data.get("model")
    values = (
        data.get("reg"),
        type_name,                                           # model
        manufacturer_from(data.get("productionLine") or type_name),
        data.get("icaoCode"),                                # ICAO type code, e.g. A320
        data.get("airlineName"),                             # owner / operator
    )
    cursor.execute(sql, values)


# Registrations used in flights but not yet in the aircraft table,
# most-used first so the most important aircraft are fetched within the cap.
missing = run_query("""
    SELECT f.aircraft_registration, COUNT(*) AS n
    FROM flights f
    LEFT JOIN aircraft a ON a.registration = f.aircraft_registration
    WHERE f.aircraft_registration IS NOT NULL AND a.registration IS NULL
    GROUP BY f.aircraft_registration
    ORDER BY n DESC
""")
aircraft_list = missing["aircraft_registration"].tolist()
print("Registrations still missing from the aircraft table:", len(aircraft_list))
aircraft_list = aircraft_list[:MAX_AIRCRAFT_CALLS]
print("Will fetch this run:", len(aircraft_list))

conn = get_connection()
cursor = conn.cursor()
success, failed = [], []

for registration in aircraft_list:
    left = units_remaining()
    if left is not None and left < MIN_UNITS_RESERVE:
        print(f"Stopping early: only {left} API units left.")
        break
    try:
        aircraft = get_aircraft_data(registration)
        if not aircraft:
            raise ValueError("no data")
        store_aircraft(cursor, aircraft)
        success.append(registration)
    except Exception as err:
        print(f"Failed : {registration} -> {err}")
        failed.append(registration)
    if len(success) % 25 == 0 and success:
        conn.commit()

conn.commit()
cursor.close()
conn.close()

print("\n" + "=" * 50)
print("VERIFYING DATA IN DATABASE")
print("=" * 50)
print(run_query("SELECT * FROM aircraft ORDER BY aircraft_id DESC LIMIT 8").to_string(index=False))
print("\nRows in aircraft table :", run_query("SELECT COUNT(*) AS n FROM aircraft")["n"][0])
print(f"Fetched this run       : {len(success)}")
print(f"Failed                 : {len(failed)}")
print(f"API units left         : {units_remaining()}")
````

## 5. Airport Delay

Two sources go into `airport_delays`: a live snapshot from the AeroDataBox delay endpoint (`api_snapshot`) and daily statistics calculated from the collected flights (`flights_calc`).

````python
def parse_hms(text):
    """'00:42:00' -> 42.0 minutes, '-00:05:00' -> -5.0 minutes."""
    if not text:
        return None
    sign = -1 if text.startswith("-") else 1
    hours, minutes, seconds = text.lstrip("-").split(":")
    return sign * (int(hours) * 60 + int(minutes) + int(seconds) / 60)


def weighted(values_weights):
    pairs = [(v, w) for v, w in values_weights if v is not None and w]
    total = sum(w for _, w in pairs)
    return sum(v * w for v, w in pairs) / total if total else None


def is_delayed(df):
    return (df["status"] == "Delayed") | (df["delay_minutes"] > DELAY_THRESHOLD_MIN)


def is_cancelled(df):
    return df["status"].fillna("").str.startswith("Cancel")


def airport_events(flights, iata):
    """One row per flight event at an airport (a departure from it or an arrival into it)."""
    dep = flights[flights["origin_iata"] == iata].assign(event_time=lambda d: d["scheduled_departure"])
    arr = flights[flights["destination_iata"] == iata].assign(event_time=lambda d: d["scheduled_arrival"])
    events = pd.concat([dep, arr], ignore_index=True)
    return events[events["event_time"].notna()].copy()


def summarise(events):
    """total, delayed, cancelled, average and median delay (early flights count as 0)."""
    delays = events["delay_minutes"].dropna().clip(lower=0)
    return {
        "total": int(len(events)),
        "delayed": int(is_delayed(events).sum()),
        "cancelled": int(is_cancelled(events).sum()),
        "avg": round(float(delays.mean()), 2) if len(delays) else None,
        "median": round(float(delays.median()), 2) if len(delays) else None,
    }


flights = run_query("""
    SELECT flight_id, origin_iata, destination_iata, scheduled_departure,
           scheduled_arrival, status, delay_minutes
    FROM flights
""")
flights["scheduled_departure"] = pd.to_datetime(flights["scheduled_departure"])
flights["scheduled_arrival"] = pd.to_datetime(flights["scheduled_arrival"])
flights["delay_minutes"] = pd.to_numeric(flights["delay_minutes"], errors="coerce")

iata_codes = run_query("SELECT iata_code FROM airport WHERE iata_code IS NOT NULL")["iata_code"].tolist()
result_rows, successful, failed = [], [], []

# --- A) Live statistics from the AeroDataBox delay endpoint (6 API units per call) ---
for iata in iata_codes:
    try:
        data = api_get(f"/airports/iata/{iata}/delays")
        if not data:
            raise ValueError("no delay data")
        dep = data.get("departuresDelayInformation") or {}
        arr = data.get("arrivalsDelayInformation") or {}
        window_from = pd.to_datetime(data["from"]["utc"], utc=True).tz_localize(None)
        window_to = pd.to_datetime(data["to"]["utc"], utc=True).tz_localize(None)

        events = airport_events(flights, iata)
        in_window = events[(events["event_time"] >= window_from) & (events["event_time"] <= window_to)]
        stats = summarise(in_window)

        total = (dep.get("numTotal") or 0) + (arr.get("numTotal") or 0)
        median = weighted([
            (parse_hms(dep.get("medianDelay")), dep.get("numQualifiedTotal")),
            (parse_hms(arr.get("medianDelay")), arr.get("numQualifiedTotal")),
        ])
        result_rows.append((
            iata, window_from.date(), "api_snapshot", total,
            min(stats["delayed"], total), stats["avg"],
            round(median, 2) if median is not None else None,
            (dep.get("numCancelled") or 0) + (arr.get("numCancelled") or 0),
        ))
        successful.append(iata)
    except Exception as err:
        print(f"Failed : {iata} -> {err}")
        failed.append(iata)

# --- B) Daily statistics calculated from the flights collected in section 3 ---
for iata in iata_codes:
    events = airport_events(flights, iata)
    if events.empty:
        continue
    events["day"] = events["event_time"].dt.date
    for day, group in events.groupby("day"):
        s = summarise(group)
        result_rows.append((iata, day, "flights_calc", s["total"], s["delayed"],
                            s["avg"], s["median"], s["cancelled"]))

insert_sql = """
INSERT INTO airport_delays
    (airport_iata, delay_date, source, total_flights, delayed_flights,
     avg_delay_min, median_delay_min, canceled_flights)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s)
ON DUPLICATE KEY UPDATE
    total_flights    = VALUES(total_flights),
    delayed_flights  = VALUES(delayed_flights),
    avg_delay_min    = VALUES(avg_delay_min),
    median_delay_min = VALUES(median_delay_min),
    canceled_flights = VALUES(canceled_flights)
"""
conn = get_connection()
cursor = conn.cursor()
cursor.executemany(insert_sql, result_rows)
conn.commit()
cursor.close()
conn.close()

print(f"Rows inserted/updated : {len(result_rows)}")
print(run_query("""
    SELECT airport_iata, SUM(total_flights) AS flights, SUM(delayed_flights) AS delayed_count,
           ROUND(100 * SUM(delayed_flights) / SUM(total_flights), 1) AS delayed_pct,
           SUM(canceled_flights) AS cancelled
    FROM airport_delays WHERE source = 'flights_calc'
    GROUP BY airport_iata ORDER BY delayed_pct DESC
""").to_string(index=False))
print(f"\nAPI snapshots fetched : {len(successful)} / {len(iata_codes)}  failed: {failed}")
print(f"API units left        : {units_remaining()}")
````

````python
# A flight is "Delayed" if the API says so or it ran more than 15 minutes late.
DELAYED = "(status = 'Delayed' OR delay_minutes > 15)"
CANCELLED = "(status LIKE 'Cancel%')"

queries = [
    (1, "Total number of flights for each aircraft model", """
        SELECT aircraft_model, COUNT(*) AS number_of_flights
        FROM flights
        WHERE aircraft_model IS NOT NULL
        GROUP BY aircraft_model
        ORDER BY number_of_flights DESC
    """),
    (2, "Aircraft assigned to more than 5 flights", """
        SELECT f.aircraft_registration AS registration,
               COALESCE(a.model, MAX(f.aircraft_model)) AS model,
               COUNT(*) AS number_of_flights
        FROM flights f
        LEFT JOIN aircraft a ON a.registration = f.aircraft_registration
        WHERE f.aircraft_registration IS NOT NULL
        GROUP BY f.aircraft_registration, a.model
        HAVING COUNT(*) > 5
        ORDER BY number_of_flights DESC
    """),
    (3, "Airports with more than 5 outbound flights", """
        SELECT a.name AS airport_name, COUNT(*) AS outbound_flights
        FROM airport a
        JOIN flights f ON f.origin_iata = a.iata_code
        GROUP BY a.name
        HAVING COUNT(*) > 5
        ORDER BY outbound_flights DESC
    """),
    (4, "Top 3 destination airports by arriving flights", """
        SELECT a.name AS airport_name, a.city, COUNT(*) AS arriving_flights
        FROM flights f
        JOIN airport a ON a.iata_code = f.destination_iata
        GROUP BY a.name, a.city
        ORDER BY arriving_flights DESC
        LIMIT 3
    """),
    (5, "Domestic or International flights", """
        SELECT flight_number, origin_iata AS origin, destination_iata AS destination,
               CASE WHEN origin_country_code = destination_country_code
                    THEN 'Domestic' ELSE 'International' END AS flight_type
        FROM flights
        WHERE origin_country_code IS NOT NULL AND destination_country_code IS NOT NULL
        ORDER BY flight_number
    """),
    (6, "Five most recent arrivals at DEL", """
        SELECT flight_number,
               COALESCE(aircraft_registration, aircraft_model) AS aircraft,
               origin_airport_name AS departure_airport,
               COALESCE(actual_arrival, scheduled_arrival) AS arrival_time
        FROM flights
        WHERE destination_iata = 'DEL' AND scheduled_arrival IS NOT NULL
        ORDER BY arrival_time DESC
        LIMIT 5
    """),
    (7, "Airports with no arriving flights", """
        SELECT a.iata_code, a.name, a.city
        FROM airport a
        LEFT JOIN flights f ON f.destination_iata = a.iata_code
        WHERE f.flight_id IS NULL
    """),
    (8, "Flights by airline and status (On Time / Delayed / Cancelled)", f"""
        SELECT airline_code,
               SUM(CASE WHEN {CANCELLED} THEN 1 ELSE 0 END)                       AS cancelled_flights,
               SUM(CASE WHEN NOT {CANCELLED} AND {DELAYED} THEN 1 ELSE 0 END)     AS delayed_flights,
               SUM(CASE WHEN NOT {CANCELLED} AND NOT COALESCE({DELAYED}, 0)
                        THEN 1 ELSE 0 END)                                         AS on_time_flights,
               COUNT(*) AS total
        FROM flights
        WHERE airline_code IS NOT NULL
        GROUP BY airline_code
        ORDER BY total DESC
    """),
    (9, "Cancelled flights with aircraft and both airports", f"""
        SELECT flight_number,
               COALESCE(aircraft_registration, aircraft_model) AS aircraft,
               origin_airport_name AS origin_airport,
               destination_airport_name AS destination_airport,
               COALESCE(scheduled_departure, scheduled_arrival) AS flight_time
        FROM flights
        WHERE {CANCELLED}
        ORDER BY flight_time DESC
    """),
    (10, "Airport pairs served by more than 2 aircraft models", """
        SELECT origin_iata, destination_iata,
               COUNT(DISTINCT aircraft_model) AS aircraft_models
        FROM flights
        WHERE aircraft_model IS NOT NULL
        GROUP BY origin_iata, destination_iata
        HAVING COUNT(DISTINCT aircraft_model) > 2
        ORDER BY aircraft_models DESC
    """),
    (11, "Percentage of delayed arrivals per destination airport", f"""
        SELECT destination_iata, MAX(destination_airport_name) AS airport,
               COUNT(*) AS arrivals,
               ROUND(100 * SUM(CASE WHEN COALESCE({DELAYED}, 0) THEN 1 ELSE 0 END) / COUNT(*), 2)
                   AS delayed_percentage
        FROM flights
        WHERE scheduled_arrival IS NOT NULL AND NOT {CANCELLED}
        GROUP BY destination_iata
        HAVING COUNT(*) >= 5
        ORDER BY delayed_percentage DESC
    """),
]

os.makedirs("query_results", exist_ok=True)
results = {}
for number, title, sql in queries:
    df_q = run_query(sql)
    results[number] = df_q
    print("\n" + "=" * 70)
    print(f"Question {number}: {title}   ({len(df_q)} rows)")
    print("=" * 70)
    print(df_q.head(10).to_string(index=False) if len(df_q) else "(no rows)")
    safe_title = "".join(ch if ch.isalnum() else "_" for ch in title)[:60]
    df_q.to_csv(f"query_results/Question{number}_{safe_title}.csv", index=False)
````

````python
%%writefile Streamlit.py
import os
import re
import textwrap
import time

import mysql.connector
import pandas as pd
import plotly.express as px
import streamlit as st

st.set_page_config(page_title="Air Tracker Dashboard", page_icon="✈️", layout="wide")

DB_CONFIG = {
    "host": "localhost",
    "user": "root",
    "password": os.getenv("DB_PASSWORD"),
    "database": "flight_analytics",
}

# A flight is "Delayed" if the API says so or it ran more than 15 minutes late.
DELAYED = "(status = 'Delayed' OR delay_minutes > 15)"
CANCELLED = "(status LIKE 'Cancel%')"


def _clean(df):
    """Turn MySQL Decimal values into floats so pandas and Plotly can use them."""
    for col in df.columns:
        if df[col].map(lambda v: type(v).__name__ == "Decimal").any():
            df[col] = pd.to_numeric(df[col].astype(float), errors="coerce")
    return df


def run_live(sql):
    """Run a SELECT straight against MySQL (no caching) and return a DataFrame."""
    conn = mysql.connector.connect(**DB_CONFIG)
    try:
        cur = conn.cursor()
        cur.execute(sql)
        df = pd.DataFrame(cur.fetchall(), columns=[d[0] for d in cur.description])
    finally:
        conn.close()
    return _clean(df)


@st.cache_data(ttl=300, show_spinner=False)
def run_query(sql):
    """Cached version, used for the page data so filters stay fast."""
    return run_live(sql)


# (number, title, sql, chart) - chart is (x column, [y columns]) or None
QUERIES = [
    (1, "Total number of flights for each aircraft model", """
        SELECT aircraft_model, COUNT(*) AS number_of_flights
        FROM flights
        WHERE aircraft_model IS NOT NULL
        GROUP BY aircraft_model
        ORDER BY number_of_flights DESC
    """, ("aircraft_model", ["number_of_flights"])),
    (2, "Aircraft assigned to more than 5 flights", """
        SELECT f.aircraft_registration AS registration,
               COALESCE(a.model, MAX(f.aircraft_model)) AS model,
               COUNT(*) AS number_of_flights
        FROM flights f
        LEFT JOIN aircraft a ON a.registration = f.aircraft_registration
        WHERE f.aircraft_registration IS NOT NULL
        GROUP BY f.aircraft_registration, a.model
        HAVING COUNT(*) > 5
        ORDER BY number_of_flights DESC
    """, ("registration", ["number_of_flights"])),
    (3, "Airports with more than 5 outbound flights", """
        SELECT a.name AS airport_name, COUNT(*) AS outbound_flights
        FROM airport a
        JOIN flights f ON f.origin_iata = a.iata_code
        GROUP BY a.name
        HAVING COUNT(*) > 5
        ORDER BY outbound_flights DESC
    """, ("airport_name", ["outbound_flights"])),
    (4, "Top 3 destination airports by arriving flights", """
        SELECT a.name AS airport_name, a.city, COUNT(*) AS arriving_flights
        FROM flights f
        JOIN airport a ON a.iata_code = f.destination_iata
        GROUP BY a.name, a.city
        ORDER BY arriving_flights DESC
        LIMIT 3
    """, ("airport_name", ["arriving_flights"])),
    (5, "Domestic or International flights", """
        SELECT flight_number, origin_iata AS origin, destination_iata AS destination,
               CASE WHEN origin_country_code = destination_country_code
                    THEN 'Domestic' ELSE 'International' END AS flight_type
        FROM flights
        WHERE origin_country_code IS NOT NULL AND destination_country_code IS NOT NULL
        ORDER BY flight_number
    """, None),
    (6, "Five most recent arrivals at DEL", """
        SELECT flight_number,
               COALESCE(aircraft_registration, aircraft_model) AS aircraft,
               origin_airport_name AS departure_airport,
               COALESCE(actual_arrival, scheduled_arrival) AS arrival_time
        FROM flights
        WHERE destination_iata = 'DEL' AND scheduled_arrival IS NOT NULL
        ORDER BY arrival_time DESC
        LIMIT 5
    """, None),
    (7, "Airports with no arriving flights", """
        SELECT a.iata_code, a.name, a.city
        FROM airport a
        LEFT JOIN flights f ON f.destination_iata = a.iata_code
        WHERE f.flight_id IS NULL
    """, None),
    (8, "Flights by airline and status (On Time / Delayed / Cancelled)", f"""
        SELECT airline_code,
               SUM(CASE WHEN {CANCELLED} THEN 1 ELSE 0 END)                    AS cancelled_flights,
               SUM(CASE WHEN NOT {CANCELLED} AND {DELAYED} THEN 1 ELSE 0 END)  AS delayed_flights,
               SUM(CASE WHEN NOT {CANCELLED} AND NOT COALESCE({DELAYED}, 0)
                        THEN 1 ELSE 0 END)                                      AS on_time_flights,
               COUNT(*) AS total
        FROM flights
        WHERE airline_code IS NOT NULL
        GROUP BY airline_code
        ORDER BY total DESC
    """, ("airline_code", ["cancelled_flights", "delayed_flights", "on_time_flights"])),
    (9, "Cancelled flights with aircraft and both airports", f"""
        SELECT flight_number,
               COALESCE(aircraft_registration, aircraft_model) AS aircraft,
               origin_airport_name AS origin_airport,
               destination_airport_name AS destination_airport,
               COALESCE(scheduled_departure, scheduled_arrival) AS flight_time
        FROM flights
        WHERE {CANCELLED}
        ORDER BY flight_time DESC
    """, None),
    (10, "Airport pairs served by more than 2 aircraft models", """
        SELECT origin_iata, destination_iata,
               COUNT(DISTINCT aircraft_model) AS aircraft_models
        FROM flights
        WHERE aircraft_model IS NOT NULL
        GROUP BY origin_iata, destination_iata
        HAVING COUNT(DISTINCT aircraft_model) > 2
        ORDER BY aircraft_models DESC
    """, ("route", ["aircraft_models"])),
    (11, "Percentage of delayed arrivals per destination airport", f"""
        SELECT destination_iata, MAX(destination_airport_name) AS airport,
               COUNT(*) AS arrivals,
               ROUND(100 * SUM(CASE WHEN COALESCE({DELAYED}, 0) THEN 1 ELSE 0 END) / COUNT(*), 2)
                   AS delayed_percentage
        FROM flights
        WHERE scheduled_arrival IS NOT NULL AND NOT {CANCELLED}
        GROUP BY destination_iata
        HAVING COUNT(*) >= 5
        ORDER BY delayed_percentage DESC
    """, ("destination_iata", ["delayed_percentage"])),
]

FORBIDDEN = re.compile(
    r"\b(insert|update|delete|drop|alter|create|truncate|grant|revoke|replace|rename|"
    r"outfile|dumpfile|load_file|call|set|lock|unlock|handler)\b",
    re.IGNORECASE,
)


def check_read_only(sql):
    """Return an error message if the text is not a single read-only SELECT."""
    text = sql.strip().rstrip(";").strip()
    if not text:
        return "Type a query first."
    if ";" in text:
        return "Only one statement at a time, please."
    if not re.match(r"^(select|with)\b", text, re.IGNORECASE):
        return "Only SELECT queries are allowed."
    if FORBIDDEN.search(text):
        return "This query contains a keyword that is not allowed (read-only mode)."
    return None


st.title("✈️ Air Tracker Dashboard")
st.markdown("### Flight Analytics & Performance Insights")

try:
    airports_df = run_query("SELECT * FROM airport ORDER BY name")
    flights_df = run_query(
        "SELECT *, COALESCE(scheduled_departure, scheduled_arrival) AS flight_time FROM flights"
    )
    delay_df = run_query(
        """
        SELECT d.airport_iata, a.name AS airport,
               SUM(d.total_flights)    AS total_flights,
               SUM(d.delayed_flights)  AS delayed_flights,
               SUM(d.canceled_flights) AS canceled_flights,
               ROUND(100 * SUM(d.delayed_flights) / NULLIF(SUM(d.total_flights), 0), 2) AS delay_pct,
               ROUND(SUM(d.avg_delay_min * d.total_flights) / NULLIF(SUM(d.total_flights), 0), 2)
                   AS avg_delay_min
        FROM airport_delays d
        JOIN airport a ON a.iata_code = d.airport_iata
        WHERE d.source = 'flights_calc'
        GROUP BY d.airport_iata, a.name
        """
    )
except mysql.connector.Error as err:
    st.error(f"❌ Database connection failed: {err}")
    st.info("Check that the MySQL service is running and the DB_PASSWORD variable is set.")
    st.stop()

st.sidebar.title("Navigation")
page = st.sidebar.radio(
    "Go To",
    [
        "Homepage Dashboard",
        "Search & Filter Flights",
        "Airport Details Viewer",
        "Delay Analysis",
        "Route Leaderboards",
        "SQL Queries",
    ],
)

if page == "Homepage Dashboard":
    st.header("Summary Statistics")

    f1, f2, f3 = st.columns(3)
    sel_airlines = f1.multiselect("Airline", sorted(flights_df["airline_code"].dropna().unique()))
    sel_status = f2.multiselect("Status", sorted(flights_df["status"].dropna().unique()))
    sel_airports = f3.multiselect("Airport", airports_df["iata_code"].tolist())

    view = flights_df
    if sel_airlines:
        view = view[view["airline_code"].isin(sel_airlines)]
    if sel_status:
        view = view[view["status"].isin(sel_status)]
    if sel_airports:
        view = view[view["origin_iata"].isin(sel_airports) | view["destination_iata"].isin(sel_airports)]

    total_flights = len(view)
    delayed = ((view["status"] == "Delayed") | (view["delay_minutes"] > 15)).sum()
    delays = view["delay_minutes"].dropna().clip(lower=0)
    avg_delay = float(delays.mean()) if len(delays) else 0.0

    c1, c2, c3, c4 = st.columns(4)
    c1.metric("Total Airports", len(airports_df))
    c2.metric("Total Flights Fetched", f"{total_flights:,}")
    c3.metric("Average Delay (mins)", round(avg_delay, 2))
    c4.metric("Delayed Flights", f"{100 * delayed / max(total_flights, 1):.1f}%")

    st.divider()
    if view.empty:
        st.warning("No flights match these filters.")
    else:
        left, right = st.columns(2)

        traffic = (
            pd.concat([view["origin_iata"], view["destination_iata"]])
            .value_counts()
            .reindex(airports_df["iata_code"])
            .fillna(0)
            .reset_index()
        )
        traffic.columns = ["Airport", "Flights"]
        left.subheader("Flights per Airport")
        left.plotly_chart(px.bar(traffic, x="Airport", y="Flights"), width="stretch")

        status_counts = view["status"].fillna("Unknown").value_counts().reset_index()
        status_counts.columns = ["Status", "Flights"]
        right.subheader("Flights by Status")
        right.plotly_chart(px.pie(status_counts, names="Status", values="Flights"), width="stretch")

elif page == "Search & Filter Flights":
    st.header("Search & Filter Flights")

    col1, col2 = st.columns(2)
    search_flight = col1.text_input("Search Flight Number")
    search_airline = col2.text_input("Search Airline Code")

    col3, col4, col5 = st.columns(3)
    status_filter = col3.selectbox(
        "Filter by Status", ["All"] + sorted(flights_df["status"].dropna().unique().tolist())
    )
    origin_filter = col4.selectbox(
        "Filter by Origin", ["All"] + sorted(flights_df["origin_iata"].dropna().unique().tolist())
    )
    destination_filter = col5.selectbox(
        "Filter by Destination", ["All"] + sorted(flights_df["destination_iata"].dropna().unique().tolist())
    )

    flight_times = pd.to_datetime(flights_df["flight_time"])
    date_range = st.date_input(
        "Filter by Date Range",
        value=(flight_times.min().date(), flight_times.max().date()),
    )

    result = flights_df.copy()
    if search_flight:
        result = result[result["flight_number"].str.contains(search_flight, case=False, na=False)]
    if search_airline:
        result = result[result["airline_code"].str.contains(search_airline, case=False, na=False)]
    if status_filter != "All":
        result = result[result["status"] == status_filter]
    if origin_filter != "All":
        result = result[result["origin_iata"] == origin_filter]
    if destination_filter != "All":
        result = result[result["destination_iata"] == destination_filter]
    if isinstance(date_range, tuple) and len(date_range) == 2:
        day = pd.to_datetime(result["flight_time"]).dt.date
        result = result[(day >= date_range[0]) & (day <= date_range[1])]

    st.write(f"**{len(result):,}** flights match")

    page_size = st.selectbox("Rows per page", [25, 50, 100, 500], index=1)
    pages = max(1, -(-len(result) // page_size))
    page_no = st.number_input("Page", min_value=1, max_value=pages, value=1, step=1)
    start = (page_no - 1) * page_size
    st.dataframe(result.iloc[start:start + page_size], width="stretch")

elif page == "Airport Details Viewer":
    st.header("Airport Details Viewer")

    labels = (airports_df["iata_code"] + " - " + airports_df["name"]).tolist()
    choice = st.selectbox("Select Airport", labels)
    iata = choice.split(" - ")[0]
    info = airports_df[airports_df["iata_code"] == iata].iloc[0]

    c1, c2, c3 = st.columns(3)
    c1.metric("City / Country", f"{info['city']}, {info['country']}")
    c2.metric("Timezone", info["timezone"])
    c3.metric("Coordinates", f"{info['latitude']:.3f}, {info['longitude']:.3f}")

    st.map(pd.DataFrame({"lat": [info["latitude"]], "lon": [info["longitude"]]}), zoom=5)

    direction = st.radio("Direction", ["All", "Departures", "Arrivals"], horizontal=True)
    linked = flights_df[
        (flights_df["origin_iata"] == iata) | (flights_df["destination_iata"] == iata)
    ]
    if direction == "Departures":
        linked = linked[linked["origin_iata"] == iata]
    elif direction == "Arrivals":
        linked = linked[linked["destination_iata"] == iata]

    st.subheader(f"Linked Flights ({len(linked):,})")
    st.dataframe(
        linked[["flight_number", "airline_code", "origin_iata", "destination_iata",
                "flight_time", "status", "delay_minutes", "aircraft_model"]].head(500),
        width="stretch",
    )

elif page == "Delay Analysis":
    st.header("Delay Analysis by Airport")

    if delay_df.empty:
        st.warning("No delay data available.")
    else:
        f1, f2 = st.columns(2)
        chosen = f1.multiselect("Airports", delay_df["airport_iata"].tolist(),
                                default=delay_df["airport_iata"].tolist())
        min_flights = f2.slider("Minimum flights", 0, int(delay_df["total_flights"].max()), 0)
        shown = delay_df[delay_df["airport_iata"].isin(chosen) & (delay_df["total_flights"] >= min_flights)]

        if shown.empty:
            st.warning("No airports match these filters.")
        else:
            c1, c2 = st.columns(2)
            c1.plotly_chart(
                px.bar(shown, x="airport_iata", y="avg_delay_min", title="Average Delay (minutes)"),
                width="stretch",
            )
            c2.plotly_chart(
                px.bar(shown, x="airport_iata", y="delay_pct", title="Delayed Flights (%)"),
                width="stretch",
            )
            st.plotly_chart(
                px.bar(shown, x="airport_iata", y="canceled_flights", title="Cancelled Flights"),
                width="stretch",
            )
            st.dataframe(shown.sort_values("delay_pct", ascending=False), width="stretch")

elif page == "Route Leaderboards":
    st.header("Route Leaderboards")

    f1, f2 = st.columns(2)
    sel_airlines = f1.multiselect("Airline", sorted(flights_df["airline_code"].dropna().unique()))
    sel_status = f2.multiselect("Status", sorted(flights_df["status"].dropna().unique()))
    route_source = flights_df.dropna(subset=["origin_iata", "destination_iata"])
    if sel_airlines:
        route_source = route_source[route_source["airline_code"].isin(sel_airlines)]
    if sel_status:
        route_source = route_source[route_source["status"].isin(sel_status)]

    st.subheader("Busiest Routes")
    routes = (
        route_source.groupby(["origin_iata", "destination_iata"])
        .size()
        .reset_index(name="Flights")
        .sort_values("Flights", ascending=False)
        .head(10)
    )
    if routes.empty:
        st.warning("No routes match these filters.")
    else:
        st.dataframe(routes, width="stretch")
        st.plotly_chart(
            px.bar(routes.assign(Route=routes["origin_iata"] + " → " + routes["destination_iata"]),
                   x="Route", y="Flights"),
            width="stretch",
        )

    st.subheader("Most Delayed Airports")
    worst = delay_df.sort_values("delay_pct", ascending=False).head(10)
    st.dataframe(worst[["airport_iata", "airport", "total_flights", "delay_pct", "avg_delay_min"]],
                 width="stretch")

elif page == "SQL Queries":
    st.header("SQL Queries")
    st.caption("Each query runs live against the MySQL database every time it is shown.")

    labels = [f"Q{n}. {title}" for n, title, _, _ in QUERIES]
    choice = st.selectbox("Choose a question", labels)
    number, title, sql, chart = QUERIES[labels.index(choice)]
    sql = textwrap.dedent(sql).strip()

    st.subheader(f"Question {number}: {title}")
    st.code(sql, language="sql")

    started = time.perf_counter()
    try:
        result = run_live(sql)
    except mysql.connector.Error as err:
        st.error(f"Query failed: {err}")
        st.stop()
    elapsed = time.perf_counter() - started

    m1, m2 = st.columns(2)
    m1.metric("Rows returned", f"{len(result):,}")
    m2.metric("Run time", f"{elapsed:.3f} s")

    if result.empty:
        st.info("The query ran successfully and returned no rows.")
    else:
        st.dataframe(result.head(1000), width="stretch")
        if len(result) > 1000:
            st.caption(f"Showing the first 1,000 of {len(result):,} rows (the download has them all).")
        st.download_button(
            "Download result as CSV",
            result.to_csv(index=False).encode("utf-8-sig"),
            file_name=f"question_{number}.csv",
            mime="text/csv",
        )

        if chart:
            x_col, y_cols = chart
            plot = result.head(15).copy()
            if x_col == "route":
                plot["route"] = plot["origin_iata"] + " → " + plot["destination_iata"]
            st.plotly_chart(
                px.bar(plot, x=x_col, y=y_cols, barmode="stack" if len(y_cols) > 1 else "relative",
                       title=f"Question {number} (top {len(plot)})"),
                width="stretch",
            )

    st.divider()
    with st.expander("Run your own SELECT query"):
        custom = st.text_area("SQL", "SELECT * FROM airport", height=120)
        if st.button("Run my query"):
            problem = check_read_only(custom)
            if problem:
                st.error(problem)
            else:
                try:
                    custom_df = run_live(custom.strip().rstrip(";"))
                    st.success(f"{len(custom_df):,} rows returned")
                    st.dataframe(custom_df.head(1000), width="stretch")
                except mysql.connector.Error as err:
                    st.error(f"Query failed: {err}")
````
