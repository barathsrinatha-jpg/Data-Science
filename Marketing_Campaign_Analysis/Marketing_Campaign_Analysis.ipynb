## 📌 Problem Statement

A retail company has run multiple marketing campaigns and recorded detailed information about customers, their demographics, spending across product categories, channel-wise purchases, and whether they accepted specific campaigns.

Management wants a consolidated analytics solution to understand who its best customers are, what drives campaign acceptance, and how different customer segments behave across products and channels.

**The task is to:**
- Clean and analyze the given marketing data.
- Build a SQL data model and write analytical queries.
- Create an interactive dashboard (Streamlit app or Power BI report) to answer key business questions.

---

## 🎯 Business Use Cases

- Which customer segments have the highest response rate to campaigns (overall and per campaign: AcceptedCmp1–5, Response)?
- How do spending patterns across products (wine, fruits, meat, fish, sweets, gold) vary by age, income, marital status, and country?
- Which channels (web, store, catalog, deals) are most used by high-value customers, and how often do they visit the website?
- Are there specific segments that are under-served (low spending, high visits, low response)?
- What are the characteristics of "ideal target customers" for future campaigns? (age range, income band, family composition, country, etc.)

## 2. Feature Engineering

```
import pandas as pd
import numpy as np
file_path = r"cleaned_marketing_data.csv"

df = pd.read_csv(file_path)
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"])
print(df["Dt_Customer"].dtype)
current_year = pd.Timestamp.today().year

df["Age"] = current_year - df["Year_Birth"]
print(df[["Year_Birth", "Age"]].head())
today = pd.Timestamp.today()

df["Customer_Tenure_Days"] = (today - df["Dt_Customer"]).dt.days
df["Customer_Tenure_Years"] = (
    df["Customer_Tenure_Days"] / 365
).round(1)
print(df[["Dt_Customer", "Customer_Tenure_Years"]].head())
df["Children"] = df["Kidhome"] + df["Teenhome"]
print(df[["Kidhome", "Teenhome", "Children"]].head())
spending_columns = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

df["Total_Spend"] = df[spending_columns].sum(axis=1)
print(df[spending_columns + ["Total_Spend"]].head())
purchase_columns = [
    "NumWebPurchases",
    "NumCatalogPurchases",
    "NumStorePurchases",
    "NumDealsPurchases"
]

df["Total_Purchases"] = df[purchase_columns].sum(axis=1)
print(df[purchase_columns + ["Total_Purchases"]].head())
campaign_columns = [
    "AcceptedCmp1",
    "AcceptedCmp2",
    "AcceptedCmp3",
    "AcceptedCmp4",
    "AcceptedCmp5"
]

df["Campaigns_Accepted"] = df[campaign_columns].sum(axis=1)
print(df[campaign_columns + ["Campaigns_Accepted"]].head())
new_columns = [
    "Age",
    "Customer_Tenure_Days",
    "Customer_Tenure_Years",
    "Children",
    "Total_Spend",
    "Total_Purchases",
    "Campaigns_Accepted"
]

print(df[new_columns].head())
output_path = r"marketing_feature_engineered.csv"

df.to_csv(output_path, index=False)
print("\n" + "=" * 60)
print("Feature Engineering Completed Successfully")
print("=" * 60)

print(f"Rows    : {df.shape[0]}")
print(f"Columns : {df.shape[1]}")

print("\nFeature-engineered dataset saved successfully.")
```

## 4. Rule-Based Segmentation

```
import pandas as pd
import numpy as np
file_path = r"marketing_feature_engineered.csv"

df = pd.read_csv(file_path)
age_bins = [0, 30, 45, 60, 100]

age_labels = [
    "Young",
    "Middle Age",
    "Senior",
    "Elder"
]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels
)
print(df["Age_Group"].value_counts())
income_bins = [
    0,
    30000,
    60000,
    90000,
    df["Income"].max()
]

income_labels = [
    "Low Income",
    "Middle Income",
    "Upper Middle",
    "High Income"
]

df["Income_Group"] = pd.cut(
    df["Income"],
    bins=income_bins,
    labels=income_labels,
    include_lowest=True
)
high_spend = df["Total_Spend"].quantile(0.90)

df["High_Spender"] = np.where(
    df["Total_Spend"] >= high_spend,
    "Yes",
    "No"
)
df["Family_Customer"] = np.where(
    df["Children"] > 0,
    "Yes",
    "No"
)
df["Campaign_Responder"] = np.where(
    df["Response"] == 1,
    "Yes",
    "No"
)
df["High_Web_Engagement"] = np.where(
    df["NumWebVisitsMonth"] > 5,
    "Yes",
    "No"
)
df["High_Income"] = np.where(
    df["Income"] > 75000,
    "Yes",
    "No"
)
q1 = df["Total_Spend"].quantile(0.25)
q3 = df["Total_Spend"].quantile(0.75)

def customer_value(spend):

    if spend >= q3:
        return "High Value"

    elif spend >= q1:
        return "Medium Value"

    else:
        return "Low Value"

df["Customer_Value"] = df["Total_Spend"].apply(customer_value)
segment_columns = [
    "Age_Group",
    "Income_Group",
    "High_Spender",
    "Family_Customer",
    "Campaign_Responder",
    "High_Web_Engagement",
    "High_Income",
    "Customer_Value"
]

print(df[segment_columns].head())
for column in segment_columns:

    print("\n" + "=" * 50)

    print(column)

    print("=" * 50)

    print(df[column].value_counts())
output_path = r"marketing_segmented.csv"

df.to_csv(output_path, index=False)

print("\nSegmentation completed successfully.")
```
