## 📌 Problem Statement

A retail company has run multiple marketing campaigns and recorded detailed information about customers, their demographics, spending across product categories, channel-wise purchases, and whether they accepted specific campaigns.

Management wants a consolidated analytics solution to understand who its best customers are, what drives campaign acceptance, and how different customer segments behave across products and channels.

**The task is to:**
- Clean and analyze the given marketing data.
- Build a SQL data model and write analytical queries.
- Create an interactive dashboard (Streamlit app or Power BI report) to answer key business questions.

---

## 🎯 Business Use Cases

- Which customer segments have the highest response rate to campaigns (overall and per campaign: AcceptedCmp1–5, Response)?
- How do spending patterns across products (wine, fruits, meat, fish, sweets, gold) vary by age, income, marital status, and country?
- Which channels (web, store, catalog, deals) are most used by high-value customers, and how often do they visit the website?
- Are there specific segments that are under-served (low spending, high visits, low response)?
- What are the characteristics of "ideal target customers" for future campaigns? (age range, income band, family composition, country, etc.)

## 0. Configuration

File names, column groups and the business rules from the brief, in one place.

```python
import os
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ---------------- file names (run the notebook from its own folder) ----------------
RAW_FILE = "marketing_campaign_data.csv"
CLEANED_FILE = "cleaned_marketing_data.csv"
FEATURE_FILE = "marketing_feature_engineered.csv"
SEGMENTED_FILE = "marketing_segmented.csv"
CHART_DIR = "charts"          # every EDA chart is saved here
TABLE_DIR = "kpi_tables"      # every KPI table is saved here
os.makedirs(CHART_DIR, exist_ok=True)
os.makedirs(TABLE_DIR, exist_ok=True)

# ---------------- column groups ----------------
SPEND_COLS = ["MntWines", "MntFruits", "MntMeatProducts",
              "MntFishProducts", "MntSweetProducts", "MntGoldProds"]
PURCHASE_COLS = ["NumWebPurchases", "NumCatalogPurchases",
                 "NumStorePurchases", "NumDealsPurchases"]
CAMPAIGN_COLS = ["AcceptedCmp1", "AcceptedCmp2", "AcceptedCmp3",
                 "AcceptedCmp4", "AcceptedCmp5"]

# ---------------- business rules from the project brief ----------------
INCOME_CAP_PERCENTILE = 0.99   # incomes above this percentile are capped (outliers)
HIGH_INCOME = 75000            # High Income      : Income > 75,000
YOUNG_AGE = 30                 # Young Customer   : Age < 30
HIGH_WEB_VISITS = 5            # High Web Engagement : NumWebVisitsMonth > 5
HIGH_SPEND_PERCENTILE = 0.90   # High Spender     : Total_Spend > 90th percentile
MARITAL_MAP = {"Absurd": "Single", "YOLO": "Single", "Alone": "Single"}


def save_chart(name):
    """Save the current figure into the charts folder and show it."""
    plt.tight_layout()
    plt.savefig(f"{CHART_DIR}/{name}.png", dpi=120)
    plt.show()
    plt.close()


def save_table(table, name):
    """Save a KPI table as CSV and return it (so it prints in the notebook)."""
    table.to_csv(f"{TABLE_DIR}/{name}.csv")
    return table


def takeaway(text):
    print(f"\n>> Takeaway: {text}\n")


print("Configuration loaded.")
```

## 1. Data Cleaning

```python
df = pd.read_csv(RAW_FILE)

print("=" * 60)
print("1. DATA UNDERSTANDING")
print("=" * 60)
print(f"Rows: {df.shape[0]:,}   Columns: {df.shape[1]}")
print("\nData types:")
print(df.dtypes.to_string())

# ---------------- missing values ----------------
print("\n" + "=" * 60)
print("2. MISSING VALUES")
print("=" * 60)
missing = df.isnull().sum()
print(missing[missing > 0].to_string() if missing.any() else "No missing values found.")

# generic safety net: numeric -> median, text -> most common value
for col in df.columns[df.isnull().any()]:
    if pd.api.types.is_numeric_dtype(df[col]):
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])

# ---------------- duplicates ----------------
print("\n" + "=" * 60)
print("3. DUPLICATES")
print("=" * 60)
before = len(df)
df = df.drop_duplicates().drop_duplicates(subset="ID")
print(f"Duplicate rows / repeated IDs removed: {before - len(df)}")

# ---------------- data types ----------------
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"])
print(f"\nDt_Customer converted to {df['Dt_Customer'].dtype}: "
      f"{df['Dt_Customer'].min().date()} to {df['Dt_Customer'].max().date()}")

# ---------------- inconsistent categories ----------------
print("\n" + "=" * 60)
print("4. CATEGORY CLEAN-UP")
print("=" * 60)
print("Marital_Status before:", df["Marital_Status"].value_counts().to_dict())
df["Marital_Status"] = df["Marital_Status"].replace(MARITAL_MAP)
print("Marital_Status after :", df["Marital_Status"].value_counts().to_dict())
print("(Absurd, YOLO and Alone are not real statuses, so they are merged into Single.)")

# ---------------- outliers and impossible values ----------------
print("\n" + "=" * 60)
print("5. OUTLIERS AND IMPOSSIBLE VALUES")
print("=" * 60)
numeric_cols = df.select_dtypes(include="number").columns
negatives = {c: int((df[c] < 0).sum()) for c in numeric_cols if (df[c] < 0).any()}
print("Columns with negative values:", negatives or "none")

enrol_year = df["Dt_Customer"].dt.year.max()
bad_birth = ((df["Year_Birth"] < 1920) | (df["Year_Birth"] > enrol_year - 18)).sum()
print(f"Unrealistic birth years (before 1920 or under 18 at enrolment): {bad_birth}")
df = df[(df["Year_Birth"] >= 1920) & (df["Year_Birth"] <= enrol_year - 18)]

income_cap = df["Income"].quantile(INCOME_CAP_PERCENTILE)
capped = int((df["Income"] > income_cap).sum())
print(f"Income: max = {df['Income'].max():,.0f}, 99th percentile = {income_cap:,.0f}")
print(f"Capped {capped} extreme incomes at {income_cap:,.0f}")
df["Income"] = df["Income"].clip(upper=income_cap)

print("\nSpending summary:")
print(df[SPEND_COLS].describe().round(1).to_string())

# ---------------- save ----------------
df.to_csv(CLEANED_FILE, index=False)
print("\n" + "=" * 60)
print(f"Cleaning complete: {len(df):,} rows, {df.shape[1]} columns -> {CLEANED_FILE}")
print("=" * 60)
```

## 2. Feature Engineering

```python
df = pd.read_csv(CLEANED_FILE, parse_dates=["Dt_Customer"])

# The data covers customers who joined in 2012-2014, so "today" for this dataset is the
# last enrolment date, NOT the real calendar date. Using the real date would make every
# customer 12 years too old and leave the "Young Customer" segment empty.
REFERENCE_DATE = df["Dt_Customer"].max()
print("Reference date for Age and Tenure:", REFERENCE_DATE.date())

# ---------------- derived columns ----------------
df["Age"] = REFERENCE_DATE.year - df["Year_Birth"]

tenure_days = (REFERENCE_DATE - df["Dt_Customer"]).dt.days
df["Customer_Tenure_Days"] = tenure_days
df["Customer_Tenure_Months"] = (tenure_days / 30.44).round(1)
df["Customer_Tenure_Years"] = (tenure_days / 365.25).round(2)

df["Children"] = df["Kidhome"] + df["Teenhome"]
df["Total_Spend"] = df[SPEND_COLS].sum(axis=1)
df["Total_Purchases"] = df[PURCHASE_COLS].sum(axis=1)
df["Campaigns_Accepted"] = df[CAMPAIGN_COLS].sum(axis=1)
df["Avg_Spend_Per_Purchase"] = (df["Total_Spend"] / df["Total_Purchases"].replace(0, np.nan)).round(2)

new_columns = ["Age", "Customer_Tenure_Days", "Customer_Tenure_Months", "Customer_Tenure_Years",
               "Children", "Total_Spend", "Total_Purchases", "Campaigns_Accepted",
               "Avg_Spend_Per_Purchase"]
print(df[new_columns].describe().round(2).to_string())

# ---------------- sanity checks ----------------
assert df["Age"].between(18, 100).all(), "unrealistic age found"
assert (df["Total_Spend"] >= 0).all()
print(f"\nAge range: {df['Age'].min()} - {df['Age'].max()}")
print(f"Tenure range: {df['Customer_Tenure_Months'].min()} - {df['Customer_Tenure_Months'].max()} months")
print(f"Customers under 30: {(df['Age'] < YOUNG_AGE).sum():,}")

df.to_csv(FEATURE_FILE, index=False)
print(f"\nFeature engineering complete: {df.shape[0]:,} rows, {df.shape[1]} columns -> {FEATURE_FILE}")
```

## 3. Exploratory Data Analysis

Every takeaway below is calculated from the data.

### 3.1 Univariate analysis

```python
df = pd.read_csv(FEATURE_FILE)
print(f"Rows: {len(df):,}   Columns: {df.shape[1]}")


def numeric_hist(col, title, bins=30, color="#4C72B0"):
    plt.figure(figsize=(9, 5))
    sns.histplot(df[col], bins=bins, kde=True, color=color)
    plt.title(title)
    plt.xlabel(col)
    plt.ylabel("Number of customers")
    save_chart(f"{col.lower()}_distribution")


def category_bar(col, title):
    counts = df[col].value_counts()
    plt.figure(figsize=(9, 5))
    sns.barplot(x=counts.index, y=counts.values, color="#55A868")
    plt.xticks(rotation=40, ha="right")
    plt.title(title)
    plt.ylabel("Number of customers")
    save_chart(f"{col.lower()}_distribution")
    top, low = counts.index[0], counts.index[-1]
    takeaway(f"{top} is the largest group ({counts.iloc[0] / len(df):.1%} of customers); "
             f"{low} is the smallest ({counts.iloc[-1] / len(df):.1%}).")


print("=" * 60)
print("UNIVARIATE ANALYSIS")
print("=" * 60)

numeric_hist("Age", "Age distribution")
takeaway(f"Median age is {df['Age'].median():.0f}; "
         f"{df['Age'].between(35, 60).mean():.0%} of customers are aged 35-60 and "
         f"{(df['Age'] < YOUNG_AGE).mean():.1%} are under {YOUNG_AGE}.")

numeric_hist("Income", "Income distribution (outliers capped)")
takeaway(f"Median income is {df['Income'].median():,.0f} against a mean of {df['Income'].mean():,.0f}; "
         f"{(df['Income'] > HIGH_INCOME).mean():.1%} of customers earn more than {HIGH_INCOME:,}.")

category_bar("Education", "Education level")
category_bar("Marital_Status", "Marital status (after clean-up)")
category_bar("Country", "Customers by country")

numeric_hist("Total_Spend", "Total spend (last 2 years)")
top10_share = df.nlargest(int(len(df) * 0.1), "Total_Spend")["Total_Spend"].sum() / df["Total_Spend"].sum()
takeaway(f"Spending is skewed: median {df['Total_Spend'].median():,.0f} vs mean {df['Total_Spend'].mean():,.0f}. "
         f"The top 10% of customers account for {top10_share:.0%} of all spend.")

numeric_hist("Total_Purchases", "Total purchases")
takeaway(f"A typical customer makes {df['Total_Purchases'].median():.0f} purchases "
         f"(middle 50%: {df['Total_Purchases'].quantile(.25):.0f}-{df['Total_Purchases'].quantile(.75):.0f}).")

numeric_hist("Recency", "Days since last purchase")
takeaway(f"Recency runs from 0 to {df['Recency'].max()} days; the median customer last bought {df['Recency'].median():.0f} days ago, "
         f"and {(df['Recency'] > 90).mean():.0%} have not bought for more than 90 days.")

# product mix
product = df[SPEND_COLS].sum().sort_values(ascending=False)
plt.figure(figsize=(9, 5))
sns.barplot(x=[c.replace("Mnt", "").replace("Prods", "").replace("Products", "") for c in product.index],
            y=product.values, color="#C44E52")
plt.title("Total spend by product category")
plt.ylabel("Total spend")
save_chart("spend_by_product")
takeaway(f"{product.index[0].replace('Mnt', '')} is the biggest category "
         f"({product.iloc[0] / product.sum():.0%} of spend), followed by {product.index[1].replace('Mnt', '')} "
         f"({product.iloc[1] / product.sum():.0%}).")

# channels
channel = df[PURCHASE_COLS].sum().sort_values(ascending=False)
plt.figure(figsize=(8, 5))
sns.barplot(x=[c.replace("Num", "").replace("Purchases", "") for c in channel.index],
            y=channel.values, color="#8172B2")
plt.title("Purchases by channel")
plt.ylabel("Number of purchases")
save_chart("purchases_by_channel")
takeaway(f"{channel.index[0].replace('Num', '').replace('Purchases', '')} is the most used channel "
         f"({channel.iloc[0] / channel.sum():.0%} of purchases).")

# campaign response
rates = (df[CAMPAIGN_COLS + ["Response"]].mean() * 100).round(1)
plt.figure(figsize=(9, 5))
sns.barplot(x=rates.index, y=rates.values, color="#DD8452")
plt.xticks(rotation=30)
plt.title("Acceptance rate by campaign (%)")
plt.ylabel("% of customers who accepted")
save_chart("campaign_acceptance_rates")
takeaway(f"The latest campaign (Response) was accepted by {rates['Response']}% of customers. "
         f"Earlier campaigns ranged from {rates[CAMPAIGN_COLS].min()}% to {rates[CAMPAIGN_COLS].max()}%.")
```

### 3.2 Bivariate analysis

```python
print("=" * 60)
print("BIVARIATE ANALYSIS")
print("=" * 60)

SAMPLE = df.sample(8000, random_state=1)   # scatter plots of 50,000+ points are unreadable


def scatter(x, y, title, name):
    plt.figure(figsize=(9, 6))
    sns.scatterplot(data=SAMPLE, x=x, y=y, alpha=0.35, s=14)
    plt.title(title)
    save_chart(name)


def box(x, y, title, name, order=None, rotate=0):
    plt.figure(figsize=(9, 5))
    sns.boxplot(data=df, x=x, y=y, order=order, showfliers=False)
    plt.xticks(rotation=rotate, ha="right" if rotate else "center")
    plt.title(title)
    save_chart(name)


scatter("Income", "Total_Spend", "Income vs total spend (8,000-customer sample)", "income_vs_total_spend")
takeaway(f"Income and spend move together (correlation {df['Income'].corr(df['Total_Spend']):.2f}), "
         "but there is wide variation at every income level.")

scatter("Age", "Total_Spend", "Age vs total spend (8,000-customer sample)", "age_vs_spend")
takeaway(f"Age has only a weak link with spend (correlation {df['Age'].corr(df['Total_Spend']):.2f}).")

box("Response", "Income", "Income by campaign response (0 = no, 1 = yes)", "income_vs_response")
inc = df.groupby("Response")["Income"].median()
takeaway(f"Responders have a much higher median income ({inc[1]:,.0f}) than non-responders ({inc[0]:,.0f}).")

box("Response", "Age", "Age by campaign response (0 = no, 1 = yes)", "age_vs_response")
age = df.groupby("Response")["Age"].mean()
takeaway(f"Responders are only slightly older on average ({age[1]:.1f} vs {age[0]:.1f} years).")

edu_order = df.groupby("Education")["Total_Spend"].median().sort_values(ascending=False).index
box("Education", "Total_Spend", "Total spend by education", "education_vs_spend", edu_order, 30)
med = df.groupby("Education")["Total_Spend"].median().sort_values(ascending=False)
takeaway(f"{med.index[0]} customers have the highest median spend ({med.iloc[0]:,.0f}); "
         f"{med.index[-1]} the lowest ({med.iloc[-1]:,.0f}).")

mar_order = df.groupby("Marital_Status")["Total_Spend"].median().sort_values(ascending=False).index
box("Marital_Status", "Total_Spend", "Total spend by marital status", "marital_status_vs_spend", mar_order, 30)
med = df.groupby("Marital_Status")["Total_Spend"].median().sort_values(ascending=False)
takeaway(f"{med.index[0]} customers spend the most (median {med.iloc[0]:,.0f}) and "
         f"{med.index[-1]} the least ({med.iloc[-1]:,.0f}).")

box("Children", "Total_Spend", "Total spend by number of children", "children_vs_spend")
med = df.groupby("Children")["Total_Spend"].median()
rising = med.get(3, 0) > med[2]
takeaway(f"Customers with no children spend the most (median {med[0]:,.0f}); with 1 child it falls to {med[1]:,.0f} "
         f"and with 2 children to {med[2]:,.0f}. "
         + ("Families with 3 or more children spend more again, so the pattern is not steadily downward."
            if rising else "Spend keeps falling as the number of children grows."))

scatter("NumWebVisitsMonth", "Total_Purchases", "Web visits per month vs total purchases", "web_visits_vs_purchases")
takeaway(f"More web visits do not mean more purchases: the correlation is "
         f"{df['NumWebVisitsMonth'].corr(df['Total_Purchases']):.2f}.")

box("Campaigns_Accepted", "Total_Spend", "Total spend by campaigns accepted", "campaigns_accepted_vs_spend")
med = df.groupby("Campaigns_Accepted")["Total_Spend"].median()
takeaway(f"Customers who accepted no earlier campaign have a median spend of {med[0]:,.0f}; "
         f"those who accepted 1 or more spend at least {med[med.index >= 1].min():,.0f}.")

plt.figure(figsize=(16, 12))
corr = df.select_dtypes(include="number").drop(columns=["ID"]).corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", annot_kws={"size": 6})
plt.title("Correlation heatmap")
save_chart("correlation_heatmap")
derived_from_spend = SPEND_COLS + ["Total_Spend", "Avg_Spend_Per_Purchase"]   # would only restate the total
strong = corr["Total_Spend"].drop(derived_from_spend).sort_values(ascending=False)
takeaway(f"Apart from the product columns themselves, Total_Spend is most strongly related to {strong.index[0]} ({strong.iloc[0]:.2f}), "
         f"{strong.index[1]} ({strong.iloc[1]:.2f}) and {strong.index[2]} ({strong.iloc[2]:.2f}).")
```

## 4. Rule-Based Segmentation

```python
df = pd.read_csv(FEATURE_FILE)

# ---------------- bands used for charts and filters ----------------
df["Age_Group"] = pd.cut(df["Age"], bins=[0, 29, 44, 59, 120],
                         labels=["Young (<30)", "Adult (30-44)", "Middle Age (45-59)", "Senior (60+)"])

# the top band starts exactly at the High Income rule (> 75,000)
df["Income_Group"] = pd.cut(df["Income"], bins=[-1, 30000, 60000, HIGH_INCOME, np.inf],
                            labels=["Low (<=30k)", "Middle (30-60k)", "Upper Middle (60-75k)", "High (>75k)"])

# ---------------- the six rule-based segments from the brief ----------------
spend_cutoff = df["Total_Spend"].quantile(HIGH_SPEND_PERCENTILE)
print(f"90th percentile of Total_Spend (High Spender cut-off): {spend_cutoff:,.0f}")

df["High_Income"] = np.where(df["Income"] > HIGH_INCOME, "Yes", "No")
df["Young_Customer"] = np.where(df["Age"] < YOUNG_AGE, "Yes", "No")
df["Campaign_Responder"] = np.where(df["Response"] == 1, "Yes", "No")
df["High_Web_Engagement"] = np.where(df["NumWebVisitsMonth"] > HIGH_WEB_VISITS, "Yes", "No")
df["Family_Customer"] = np.where(df["Children"] > 0, "Yes", "No")
df["High_Spender"] = np.where(df["Total_Spend"] > spend_cutoff, "Yes", "No")

# ---------------- customer value tiers (quartiles of total spend) ----------------
q1, q3 = df["Total_Spend"].quantile([0.25, 0.75])


def customer_value(spend):
    if spend >= q3:
        return "High Value"
    if spend >= q1:
        return "Medium Value"
    return "Low Value"


df["Customer_Value"] = df["Total_Spend"].apply(customer_value)

segment_flags = ["High_Income", "Young_Customer", "Campaign_Responder",
                 "High_Web_Engagement", "Family_Customer", "High_Spender"]
summary = pd.DataFrame({
    "Rule": ["Income > 75,000", "Age < 30", "Response = 1", "Web visits > 5",
             "Children > 0", "Total_Spend > 90th percentile"],
    "Customers": [(df[c] == "Yes").sum() for c in segment_flags],
    "Share_%": [round(100 * (df[c] == "Yes").mean(), 1) for c in segment_flags],
}, index=segment_flags)
print("\nRule-based segments:")
print(summary.to_string())

for col in ["Age_Group", "Income_Group", "Customer_Value"]:
    print(f"\n{col}:")
    print(df[col].value_counts().sort_index().to_string())

df.to_csv(SEGMENTED_FILE, index=False)
save_table(summary, "segment_sizes")
print(f"\nSegmentation complete: {df.shape[0]:,} rows, {df.shape[1]} columns -> {SEGMENTED_FILE}")
```

## 5. Business Questions

The five business use cases from the brief, answered with KPI tables and charts.

```python
df = pd.read_csv(SEGMENTED_FILE)
overall_rate = df["Response"].mean() * 100


def response_by(col):
    """Customers, responders and response rate for every value of a column."""
    out = df.groupby(col, observed=True)["Response"].agg(Customers="size", Responders="sum")
    out["Response_Rate_%"] = (100 * out["Responders"] / out["Customers"]).round(1)
    return out.sort_values("Response_Rate_%", ascending=False)


print("=" * 70)
print("Q1. WHICH SEGMENTS RESPOND MOST TO CAMPAIGNS?")
print("=" * 70)
campaign_rates = (df[CAMPAIGN_COLS + ["Response"]].mean() * 100).round(2).rename("Acceptance_Rate_%")
print(save_table(campaign_rates.to_frame(), "campaign_acceptance_rates").to_string())
print(f"\nOverall response rate (latest campaign): {overall_rate:.1f}%")

segment_rows = []
for col in ["High_Income", "Young_Customer", "High_Web_Engagement", "Family_Customer", "High_Spender"]:
    for value in ["Yes", "No"]:
        part = df[df[col] == value]
        segment_rows.append({"Segment": f"{col} = {value}", "Customers": len(part),
                             "Response_Rate_%": round(100 * part["Response"].mean(), 1)})
seg_table = pd.DataFrame(segment_rows).set_index("Segment")
print()
print(save_table(seg_table, "response_by_rule_segment").to_string())

for col in ["Income_Group", "Age_Group", "Marital_Status", "Education", "Country"]:
    table = save_table(response_by(col), f"response_by_{col.lower()}")
    print(f"\nResponse by {col}:")
    print(table.to_string())

best_income = response_by("Income_Group").iloc[0]
fam = seg_table["Response_Rate_%"]
takeaway(f"High earners respond best: the '{response_by('Income_Group').index[0]}' band has a "
         f"{best_income['Response_Rate_%']}% response rate against {overall_rate:.1f}% overall. "
         f"Customers without children respond at {fam['Family_Customer = No']}% vs "
         f"{fam['Family_Customer = Yes']}% for families.")

plt.figure(figsize=(10, 5))
plot = pd.concat([seg_table.loc[[f"{c} = Yes" for c in
                                 ["High_Income", "Young_Customer", "High_Web_Engagement",
                                  "Family_Customer", "High_Spender"]]]])
sns.barplot(x=plot["Response_Rate_%"], y=plot.index, color="#4C72B0")
plt.axvline(overall_rate, color="red", linestyle="--", label=f"Overall {overall_rate:.1f}%")
plt.legend()
plt.title("Response rate of each rule-based segment")
save_chart("response_by_segment")

print("\n" + "=" * 70)
print("Q2. HOW DOES PRODUCT SPENDING VARY BY AGE, INCOME, MARITAL STATUS AND COUNTRY?")
print("=" * 70)
for col in ["Age_Group", "Income_Group", "Marital_Status", "Country"]:
    table = save_table(df.groupby(col, observed=True)[SPEND_COLS].mean().round(1), f"avg_spend_by_{col.lower()}")
    print(f"\nAverage spend per customer by {col}:")
    print(table.to_string())

plt.figure(figsize=(11, 5))
sns.heatmap(df.groupby("Country")[SPEND_COLS].mean(), annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Average spend per customer: country x product")
save_chart("spend_country_by_product")
country_total = df.groupby("Country")["Total_Spend"].mean().sort_values(ascending=False)
prod_rank = df[SPEND_COLS].sum().sort_values(ascending=False)
takeaway(f"{prod_rank.index[0].replace('Mnt', '')} and {prod_rank.index[1].replace('Mnt', '')} make up "
         f"{prod_rank.iloc[:2].sum() / prod_rank.sum():.0%} of all spend. {country_total.index[0]} customers spend the most "
         f"on average ({country_total.iloc[0]:,.0f}) and {country_total.index[-1]} the least ({country_total.iloc[-1]:,.0f}).")

print("\n" + "=" * 70)
print("Q3. WHICH CHANNELS DO HIGH-VALUE CUSTOMERS USE, AND HOW OFTEN DO THEY VISIT THE WEBSITE?")
print("=" * 70)
channel_cols = PURCHASE_COLS + ["NumWebVisitsMonth"]
channels = save_table(df.groupby("Customer_Value")[channel_cols].mean().round(2).loc[["High Value", "Medium Value", "Low Value"]],
                      "channel_use_by_customer_value")
print(channels.to_string())
channels.drop(columns="NumWebVisitsMonth").plot(kind="bar", figsize=(9, 5), rot=0)
plt.title("Average purchases per customer by channel and customer value")
plt.ylabel("Average purchases")
save_chart("channels_by_customer_value")
hv = channels.loc["High Value"]
top_ch = hv[PURCHASE_COLS].sort_values(ascending=False)
name = lambda c: c.replace("Num", "").replace("Purchases", "").lower()
takeaway(f"High-value customers buy mostly through the {name(top_ch.index[0])} channel ({top_ch.iloc[0]:.1f} purchases per "
         f"customer) and {name(top_ch.index[1])} ({top_ch.iloc[1]:.1f}), and visit the website {hv['NumWebVisitsMonth']:.1f} "
         f"times a month (low-value customers: {channels.loc['Low Value', 'NumWebVisitsMonth']:.1f}).")

print("\n" + "=" * 70)
print("Q4. WHICH SEGMENT IS UNDER-SERVED? (low spend + high web visits + did not respond)")
print("=" * 70)
under = df[(df["Total_Spend"] < df["Total_Spend"].median())
           & (df["NumWebVisitsMonth"] > HIGH_WEB_VISITS)
           & (df["Response"] == 0)]
print(f"Under-served customers: {len(under):,} ({len(under) / len(df):.1%} of all customers)")
profile = pd.DataFrame({
    "Under_served": [under["Age"].mean(), under["Income"].mean(), under["Children"].mean(), under["Total_Spend"].mean(),
                     under["NumWebVisitsMonth"].mean()],
    "All_customers": [df["Age"].mean(), df["Income"].mean(), df["Children"].mean(), df["Total_Spend"].mean(),
                      df["NumWebVisitsMonth"].mean()],
}, index=["Avg_Age", "Avg_Income", "Avg_Children", "Avg_Total_Spend", "Avg_Web_Visits"]).round(1)
print(save_table(profile, "under_served_profile").to_string())
print("\nTop countries:", under["Country"].value_counts().head(3).to_dict())
takeaway(f"{len(under):,} customers visit the website often but spend little and did not respond to the campaign. "
         f"They earn {under['Income'].mean():,.0f} on average (all customers: {df['Income'].mean():,.0f}) and have "
         f"{under['Children'].mean():.1f} children (all customers: {df['Children'].mean():.1f}). "
         "This group is the first candidate for a targeted, offer-led campaign.")

print("\n" + "=" * 70)
print("Q5. WHAT DOES THE IDEAL TARGET CUSTOMER LOOK LIKE?")
print("=" * 70)
compare = df.groupby("Response")[["Age", "Income", "Children", "Total_Spend", "Total_Purchases",
                                  "NumWebVisitsMonth", "Recency"]].mean().round(1).T
compare.columns = ["Non-responders", "Responders"]
print(save_table(compare, "responder_profile").to_string())

combos = (df.groupby(["Income_Group", "Family_Customer", "Age_Group"], observed=True)["Response"]
          .agg(Customers="size", Response_Rate=lambda s: round(100 * s.mean(), 1))
          .query("Customers >= 300").sort_values("Response_Rate", ascending=False).head(5))
print("\nBest-responding profiles (at least 300 customers):")
print(save_table(combos, "best_responding_profiles").to_string())
top = combos.index[0]
takeaway(f"The ideal target is: income band '{top[0]}', family customer = {top[1]}, age group '{top[2]}' "
         f"- {combos.iloc[0]['Response_Rate']}% respond ({int(combos.iloc[0]['Customers']):,} customers). "
         f"Responders earn {compare.loc['Income', 'Responders']:,.0f} vs {compare.loc['Income', 'Non-responders']:,.0f}, "
         f"have {compare.loc['Children', 'Responders']} children vs {compare.loc['Children', 'Non-responders']}, and "
         f"spend {compare.loc['Total_Spend', 'Responders']:,.0f} vs {compare.loc['Total_Spend', 'Non-responders']:,.0f}.")
```

## 6. Load to MySQL

```python
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

DB_NAME = "marketing_campaign_analysis"
password = os.getenv("DB_PASSWORD")        # set once with:  setx DB_PASSWORD "your_password"

df = pd.read_csv(SEGMENTED_FILE, parse_dates=["Dt_Customer"])

server_url = URL.create("mysql+pymysql", username="root", password=password, host="localhost", port=3306)
engine = create_engine(server_url)
with engine.begin() as conn:
    conn.execute(text(f"CREATE DATABASE IF NOT EXISTS {DB_NAME}"))
engine = create_engine(server_url.set(database=DB_NAME))
print("Connected successfully")

# The table is rebuilt from the cleaned data every time, so re-running never duplicates rows.
# (Views that read from customers keep working because every old column name is kept.)
DDL = """
CREATE TABLE customers (
    ID INT PRIMARY KEY,
    Year_Birth INT,
    Education VARCHAR(50),
    Marital_Status VARCHAR(50),
    Income DECIMAL(12,2),
    Kidhome INT,
    Teenhome INT,
    Dt_Customer DATE,
    Recency INT,
    MntWines INT,
    MntFruits INT,
    MntMeatProducts INT,
    MntFishProducts INT,
    MntSweetProducts INT,
    MntGoldProds INT,
    NumDealsPurchases INT,
    NumWebPurchases INT,
    NumCatalogPurchases INT,
    NumStorePurchases INT,
    NumWebVisitsMonth INT,
    AcceptedCmp1 INT,
    AcceptedCmp2 INT,
    AcceptedCmp3 INT,
    AcceptedCmp4 INT,
    AcceptedCmp5 INT,
    Response INT,
    Complain INT,
    Country VARCHAR(50),
    Age INT,
    Customer_Tenure_Days INT,
    Customer_Tenure_Months DECIMAL(6,1),
    Customer_Tenure_Years DECIMAL(6,2),
    Children INT,
    Total_Spend DECIMAL(12,2),
    Total_Purchases INT,
    Campaigns_Accepted INT,
    Avg_Spend_Per_Purchase DECIMAL(12,2),
    Age_Group VARCHAR(30),
    Income_Group VARCHAR(30),
    High_Spender VARCHAR(10),
    Young_Customer VARCHAR(10),
    Family_Customer VARCHAR(10),
    Campaign_Responder VARCHAR(10),
    High_Web_Engagement VARCHAR(10),
    High_Income VARCHAR(10),
    Customer_Value VARCHAR(30),
    INDEX idx_country (Country),
    INDEX idx_income_group (Income_Group),
    INDEX idx_age_group (Age_Group)
)
"""
with engine.begin() as conn:
    conn.execute(text("DROP TABLE IF EXISTS customers"))
    conn.execute(text(DDL))
print("Table 'customers' created")

columns = pd.read_sql(text("SELECT COLUMN_NAME FROM information_schema.COLUMNS "
                           "WHERE TABLE_SCHEMA = :db AND TABLE_NAME = 'customers' "
                           "ORDER BY ORDINAL_POSITION"), engine, params={"db": DB_NAME})["COLUMN_NAME"].tolist()
df["Dt_Customer"] = df["Dt_Customer"].dt.date
df[columns].to_sql(name="customers", con=engine, if_exists="append", index=False, chunksize=5000)
print(f"Data uploaded successfully: {len(df):,} rows")

check = pd.read_sql(text("""
    SELECT COUNT(*) AS customers,
           MIN(Age) AS min_age, MAX(Age) AS max_age,
           SUM(Young_Customer = 'Yes') AS young_customers,
           ROUND(100 * AVG(Response), 1) AS response_rate_pct
    FROM customers
"""), engine)
print("\nVerification from MySQL:")
print(check.to_string(index=False))
```
