```python
import pandas as pd
from nltk.tokenize import word_tokenize
import unicodedata
import re
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
import torch
from torch import nn,optim
import numpy as np
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
s_w = stopwords.words('english')
def clean_text(text):
    text = str(text).lower()
    text = text.strip()
    text = unicodedata.normalize("NFKC",text)
    #http://,https://,www,
    text = re.sub(r"http://\S+|https://\S+|www\S+","",text) # remove urls
    text = re.sub(r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9._%+-]+.[A-Za-z0-9._%+-]+","",text)# removing emails
    text = re.sub(r"@\w+","",text)
    text = re.sub(r"#\w+","",text)
    text = re.sub(r"[^A-Za-z\s]+","",text)
    tokens = word_tokenize(text)
    cleaned_tokens = []
    #stemmer = PorterStemmer()
    lemmatizer = WordNetLemmatizer()
    for token in tokens:
        if (token not in s_w) and (len(token)>2):
            #token = stemmer.stem(token)
            token = lemmatizer.lemmatize(token)
            cleaned_tokens.append(token)

    cleaned_text = " ".join(cleaned_tokens)
    return cleaned_text
df = pd.read_csv(r"C:\Users\barat\OneDrive\Documents\GitHub\Data-Science\Data\Corona_NLP_test.csv")

df = df[['OriginalTweet','Sentiment']]
df['tweet'] = df['OriginalTweet'].apply(clean_text)

PAD_TOKEN = "<PAD>"
UNK_TOKEN = "<UNK>"

word_to_index = {
    PAD_TOKEN: 0,
    UNK_TOKEN: 1
}
for sentence in df["tweet"]:

    for word in sentence.split():

        if word not in word_to_index:
            word_to_index[word] = len(word_to_index)
def text_to_sequence(text):

    return [
        word_to_index.get(word, word_to_index[UNK_TOKEN])
        for word in text.split()
    ]


sequences = df["tweet"].apply(text_to_sequence).tolist()
max_length = max(len(sequence) for sequence in sequences)
def pad_sequence(sequence, max_length):

    # Truncate long sequences
    sequence = sequence[:max_length]

    # Add padding
    sequence = sequence + [word_to_index[PAD_TOKEN]] * (
        max_length - len(sequence)
    )

    return sequence


X = np.array(
    [
        pad_sequence(sequence, max_length)
        for sequence in sequences
    ]
)
X = torch.tensor(X, dtype=torch.long)
label_to_index = {
    label: index
    for index, label in enumerate(
        sorted(df["Sentiment"].unique())
    )
}

y = torch.tensor(
    [
        label_to_index[label]
        for label in df["Sentiment"]
    ],
    dtype=torch.long
)



class TextProcessing(nn.Module):
    def __init__(self,vocab_size,embedding_dim,hidden_dim,output,padding_idx=0):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size,embedding_dim,padding_idx=padding_idx)
        self.rnn = nn.RNN(embedding_dim,hidden_dim,batch_first=True)
        self.fc1 = nn.Linear(hidden_dim,output)
    def forward(self,X):
        X = self.embedding(X)
        output, hidden = self.rnn(X)
        hidden = hidden[-1]
        X = self.fc1(hidden)
        return X
vocab_size = len(word_to_index)

embedding_dim = 128

hidden_dim = 64

output = len(label_to_index)
 
model = TextProcessing(vocab_size,embedding_dim,hidden_dim,output)
optimizer = optim.Adam(model.parameters())
criterion = nn.CrossEntropyLoss()

num_epoch = 10000
for epoch in range(1,num_epoch+1):
    pred = model(X)
    loss = criterion(pred,y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if epoch%100==0:
        print(f"{epoch}/{num_epoch}--> Loss--> {loss.item()}")
pred = model(X)
pred = torch.argmax(pred,dim=1)

print(f"Accuracy: {accuracy_score(y,pred)}")
print(f"PRecision: {precision_score(y,pred,average="micro")}")
print(f"Recall: {recall_score(y,pred,average="micro")}")
print(f"F1 Score: {f1_score(y,pred,average="micro")}")
```
