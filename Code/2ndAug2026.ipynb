```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from torch import nn,optim
import torch
from torchvision import datasets,transforms
from torch.utils.data import DataLoader
transformation = transforms.Compose([
    transforms.Resize((28,28)),
    transforms.ToTensor()
    
])

train_data = datasets.MNIST(root="C:\\Users\\barat\\OneDrive\\Documents\\GitHub\\Data-Science\\Data\\MNIST",train=True,download=True,transform=transformation)
test_data = datasets.MNIST(root="C:\\Users\\barat\\OneDrive\\Documents\\GitHub\\Data-Science\\Data\\MNIST",train=False,download=True,transform=transformation)

train_loader =DataLoader(train_data,batch_size=32,shuffle=True)
test_loader = DataLoader(test_data,batch_size=64,shuffle=True)
class SimpleCNN(nn.Module):
    def __init__(self):
        super(SimpleCNN,self).__init__()
        self.conv1= nn.Conv2d(in_channels=1,out_channels=8,kernel_size=5,stride=2,padding=1)
        self.pool1 = nn.AvgPool2d(kernel_size=2,stride=2)
        self.conv2 = nn.Conv2d(in_channels=8,out_channels=16,kernel_size=3,padding=2,stride=2)
        self.pool2 = nn.AvgPool2d(kernel_size=2,stride=1)

        self.flatten = nn.Flatten() # convert 2d img to 1D
        self.fc1 = nn.Linear(144,32)
        self.fc2 = nn.Linear(32,10)
        
        self.relu = nn.ReLU()
    def forward(self,X):
        X = self.relu (self.conv1(X))
        X = self.pool1(X)
        X = self. relu (self.conv2(X))
        X = self.pool2(X)

        X = self. flatten(X)
        X = self.relu (self.fc1(X))
        X = self.fc2(X)
        return X

model = SimpleCNN()
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters())
num_epoch = 100
last_loss = torch.inf
for epoch in range(1,num_epoch+1):
    total_loss = 0
    optimizer.zero_grad()
    for x,y in train_loader:
        pred = model(x)
        loss = criterion(pred,y)
        loss.backward()
        optimizer.step()
        total_loss+=loss.item()
    
    print(f"{epoch}/{num_epoch}--> Loss--> {total_loss}")
    if last_loss<total_loss:
        break
    last_loss = total_loss

model.eval()

with torch.no_grad():
    train_pred = []
    y_train = []
    test_pred = []
    y_test = []
    for x,y in train_loader:
        pred = model(x)
        pred = torch.argmax(pred,dim=1)
        train_pred.extend(pred)
        y_train.extend(y)
    for x,y in test_loader:
        pred = model(x)
        pred = torch.argmax(pred,dim=1)
        test_pred.extend(pred)
        y_test.extend(y)
    print(f"*"*30)
    print(f"Accuracy: {accuracy_score(y_train,train_pred)}")
    print(f"Precision: {precision_score(y_train,train_pred,average="micro")}")
    print(f"Recall: {recall_score(y_train,train_pred,average="micro")}")
    print(f"F1 Score: {f1_score(y_train,train_pred,average="micro")}")

    print(f"*"*30)
    print(f"Accuracy: {accuracy_score(y_test,test_pred)}")
    print(f"Precision: {precision_score(y_test,test_pred,average="micro")}")
    print(f"Recall: {recall_score(y_test,test_pred,average="micro")}")
    print(f"F1 Score: {f1_score(y_test,test_pred,average="micro")}")
```
