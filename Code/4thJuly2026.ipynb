```python
from imblearn.over_sampling import RandomOverSampler
import seaborn as sns
import pandas as pd

df = sns.load_dataset("titanic")
df.drop(['class','embark_town','deck','alive'], axis=1, inplace=True)

df.dropna(inplace=True)
df=pd.get_dummies(df,columns=['sex','embarked','who','adult_male','alone'],dtype='int',drop_first=True)
X=df.drop('survived',axis=1)
y=df['survived']
```

```python
sampler = RandomOverSampler()
new_x,new_y= sampler.fit_resample(X,y)
len(X),len(new_x)
```

```python
new_y.value_counts()
```

```python
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
import matplotlib.pyplot as plt
X=df.drop('survived',axis=1)
y=df['survived']
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)

models = [
    DecisionTreeClassifier(),
    RandomForestClassifier(),
    GradientBoostingClassifier()
]

for model in models:

    model.fit(x_train, y_train)

    train_predictions = model.predict(x_train)
    test_predictions = model.predict(x_test)

    print("-" * 60)
    print(model.__class__.__name__)
    print("-" * 60)

    print(f"Train Accuracy : {accuracy_score(y_train, train_predictions):.4f}")
    print(f"Test Accuracy  : {accuracy_score(y_test, test_predictions):.4f}")

    print(f"Train Precision: {precision_score(y_train, train_predictions):.4f}")
    print(f"Test Precision : {precision_score(y_test, test_predictions):.4f}")

    print(f"Train Recall   : {recall_score(y_train, train_predictions):.4f}")
    print(f"Test Recall    : {recall_score(y_test, test_predictions):.4f}")

    print(f"Train F1 Score : {f1_score(y_train, train_predictions):.4f}")
    print(f"Test F1 Score  : {f1_score(y_test, test_predictions):.4f}")
```
