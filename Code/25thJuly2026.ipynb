```python
import pandas as pd
import seaborn as sns
import torch
from torch import nn,optim
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score


df = sns.load_dataset('titanic')
df.drop(['class','deck','embark_town','alive'],axis=1,inplace=True)
df.dropna(inplace=True)
df=pd.get_dummies(df,columns=['sex','embarked','who','adult_male','alone'],dtype=int,drop_first=True)
X=torch.tensor(df.drop('survived',axis=1).values,dtype=torch.float32)
y=torch.tensor(df['survived'].values,dtype=torch.float32).view(-1,1)

x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.15)

class SimpleNN(nn.Module):
    def __init__(self,input,output):
        super().__init__()
        self.fc1=nn.Linear(input,32)
        self.fc2=nn.Linear(32,16)
        self.fc3=nn.Linear(16,8)
        self.fc4=nn.Linear(8,4)
        self.fc5=nn.Linear(4,output)
        self.sigmoid=nn.Sigmoid()
        self.relu=nn.ReLU()
        self.tanh = nn.Tanh()

    def forward(self,X):
        X=self.tanh(self.fc1(X))
        X=self.relu(self.fc2(X))
        X=self.tanh(self.fc3(X))
        X=self.relu(self.fc4(X))
        X=self.relu(self.fc5(X))
        return X

model = SimpleNN(input=X.shape[1],output=1)   
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters())
num_epochs = 10000
for epoch in range(1,num_epochs+1):
    optimizer.zero_grad()
    output = model(x_train)
    loss = criterion(torch.sigmoid(output),y_train)
    loss.backward()
    optimizer.step()
    
    if epoch % 100 == 0:
        print(f'Epoch {epoch}/{num_epochs}, Loss: {loss.item()}')
model.eval()
with torch.no_grad():
    train_output = model(x_train)
    train_output = (train_output>=0.5).float()
    test_output = model(x_test)
    test_output = (test_output>=0.5).float()

    print('-' * 12)
    print("Train Accuracy:", accuracy_score(y_train, train_output))
    print("Train Precision:", precision_score(y_train, train_output))
    print("Train Recall:", recall_score(y_train, train_output))
    print("Train F1 Score:", f1_score(y_train, train_output))
    print("Test F1 Score:", f1_score(y_test, test_output))

    print('-' * 12)
    print("Test Accuracy:", accuracy_score(y_test, test_output))
    print("Test Precision:", precision_score(y_test, test_output))
    print("Test Recall:", recall_score(y_test, test_output))
    print("Test F1 Score:", f1_score(y_test, test_output))
```
