```python
df.dropna(inplace=True)
df = pd.get_dummies(df, columns=['fuel_type'],dtype='int', drop_first=True)
df
```

```python
df['kms_driven'] = pd.to_numeric(df['kms_driven'].replace({'kms': '', ' ': '',",": ""}, regex=True))
df['Price'] = pd.to_numeric(df['Price'].replace({'Ask For Price': '', ' ': '',",": ""}, regex=True))
df.dropna(inplace=True)
df
```
