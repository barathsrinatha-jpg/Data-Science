```python
import pandas as pd
from apyori import apriori

df = pd.read_csv(r"C:\Users\barat\Downloads\Market_Basket_Optimisation.csv",header=None)
all_transactions = []
for i in df.index:
    row = df.loc[i].values
    transaction = []
    for item in row:
        if str(item)!="nan":
            transaction.append(item)
    all_transactions.append(transaction)

result = pd.DataFrame(list(apriori(all_transactions,min_support = 0.003,min_confidence=0.03,min_lift=0.2)))
support = result['support']
first_value = []
second_value=[]
third_value = []
fourth_value = []
for i in result.index:
    row = result['ordered_statistics'].loc[i][0]
    first_value.append(list(row[0]))
    second_value.append(list(row[1]))
    third_value.append(row[2])
    fourth_value.append(row[3])
lhs = pd.DataFrame(first_value) 
rhs = pd.DataFrame(second_value)
confidence = pd.DataFrame(third_value,columns=['Confidence'])
lift = pd.DataFrame(fourth_value,columns=['Lift'])
output = pd.concat([lhs,rhs,support,confidence,lift],axis=1)
output.sort_values(by=['Lift'],ascending=False)
```
