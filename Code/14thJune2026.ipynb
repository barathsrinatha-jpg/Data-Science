```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LinearRegression
df = pd.read_csv(r"C:\Users\barat\Downloads\quikr_car.csv")
df.dropna(inplace=True)
df = pd.get_dummies(df, columns=['fuel_type'],dtype='int', drop_first=True)
df['kms_driven'] = pd.to_numeric(df['kms_driven'].replace({'kms': '', ' ': '',",": ""}, regex=True))
df['Price'] = pd.to_numeric(df['Price'].replace({'Ask For Price': '', ' ': '',",": ""}, regex=True))
df.dropna(inplace=True)
df = df[["Price","kms_driven"]]
df
```

```python
sns.scatterplot(x='kms_driven',y='Price',data=df)
```

```python
df = df[(df["Price"]<df["Price"].max()) & (df["kms_driven"]<df["kms_driven"].max())]
sns.scatterplot(x='kms_driven',y='Price',data=df)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
x=df.drop(["Price"], axis=1)
y=df["Price"]
model.fit(x,y)
df["predicted_price"] = model.predict(x)
plt.figure(figsize=(10,5))
sns.scatterplot(x='kms_driven',y='Price',data=df)
sns.lineplot(x=df["kms_driven"],y=df["predicted_price"],color='red')
plt.show()
df["error"] = df["Price"] - df["predicted_price"]
df["error_squared"] = df["error"]**2
df["error_squared"].mean()
```
