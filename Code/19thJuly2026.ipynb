```python
import seaborn as sns
import pandas as pd
from torch import nn,optim
from sklearn.metrics import accuracy_score

df = sns.load_dataset('titanic')
df.drop(["class", "embarked", "deck","alive"], axis=1, inplace=True)
df.dropna(inplace=True)
df = pd.get_dummies(df, columns=["sex", "who", "adult_male", "embark_town", "alone"], dtype="int", drop_first=True)
X=torch.tensor(df.drop("survived",axis=1).values, dtype=torch.float)
y=torch.tensor(df["survived"].values, dtype=torch.float).view(-1, 1)

class SimpleNN(nn.Module):
    def __init__(self,input,output):
        super().__init__() # Call nn.Module constructor
        self.fc1=nn.Linear(input,16) #input layer and hidden layer 1
        self.fc2=nn.Linear(16,32) # hidden layer 1 and hidden layer 2
        self.fc3=nn.Linear(32,8) # hidden layer 2 and hidden layer 3
        self.fc4=nn.Linear(8,output) # hidden layer 3 and output layer
        self.sigmoid=nn.Sigmoid() # sigmoid function
    def forward(self,X):
        X=self.fc1(X)
        X=self.fc2(X)
        X=self.fc3(X)
        X=self.fc4(X)
        X=self.sigmoid(X)
        return X
model=SimpleNN(X.shape[1],1)
criterion=nn.BCELoss()
optimizer=optim.Adam(model.parameters())
num_epochs=10000
for epoch in range(1,num_epochs+1):
    optimizer.zero_grad() # reset gradients
    pred = model(X) # Predict output from first round
    loss = criterion(pred,y) # calculate lose in a given epoch
    loss.backward() # perform backward propagation
    optimizer.step() #update weights
    output = (pred>0.5).float()
    if epoch%100==0:
        print(f"Epoch{epoch}/{num_epochs},Loss:{loss.item()},Accuracy:{accuracy_score(y,output)}")
```
