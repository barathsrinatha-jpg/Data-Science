```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,precision_score, recall_score, f1_score, confusion_matrix
df = pd.read_csv(r"C:\Users\barat\Downloads\Teen_Mental_Health_Dataset.csv")
df = pd.get_dummies(df, columns=["gender","platform_usage","social_interaction_level"], dtype="int", drop_first=True)
df.dropna(inplace=True)
X=df.drop("depression_label", axis=1)
y=df["depression_label"]
x_train, x_test, y_train, y_test = train_test_split(X,y,test_size=0.15)
model=KNeighborsClassifier(n_neighbors=3)
model.fit(x_train,y_train)
train_predictions=model.predict(x_train)
test_predictions=model.predict(x_test)
print(f"Train")
print(f"Train Accuracy: {accuracy_score(y_train, train_predictions)}")
print(f"Train Precision: {precision_score(y_train, train_predictions)}")
print(f"Train Recall: {recall_score(y_train, train_predictions)}")
print(f"Train F1 Score: {f1_score(y_train, train_predictions)}")
print(f"Test")
print(f"Test Accuracy: {accuracy_score(y_test, test_predictions)}")
print(f"Test Precision: {precision_score(y_test, test_predictions)}")
print(f"Test Recall: {recall_score(y_test, test_predictions)}")
print(f"Test F1 Score: {f1_score(y_test, test_predictions)}")
```

```python
df = pd.read_csv(r"C:\Users\barat\Downloads\Obesity prediction.csv")
df=pd.get_dummies(df, columns=["Gender","family_history","FAVC","CAEC","SMOKE","SCC","CALC","MTRANS"], dtype="int", drop_first=True)
X=df.drop("Obesity", axis=1)
y=df["Obesity"]
x_train, x_test, y_train, y_test = train_test_split(X,y,test_size=0.20)
model=KNeighborsClassifier(n_neighbors=3)
model.fit(x_train,y_train)
train_predictions=model.predict(x_train)
test_predictions=model.predict(x_test)
print(f"Train")
print(f"Train Accuracy: {accuracy_score(y_train, train_predictions)}")
print(f"Train Precision: {precision_score(y_train, train_predictions,average='micro')}")
print(f"Train Recall: {recall_score(y_train, train_predictions,average='micro')}")
print(f"Train F1 Score: {f1_score(y_train, train_predictions,average='micro')}")
print(f"Test")
print(f"Test Accuracy: {accuracy_score(y_test, test_predictions)}")
print(f"Test Precision: {precision_score(y_test, test_predictions,average='micro')}")
print(f"Test Recall: {recall_score(y_test, test_predictions,average='micro')}")
print(f"Test F1 Score: {f1_score(y_test, test_predictions,average='micro')}")
```
