```python
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")
X = df.drop(["species"],axis=1)
y= df["species"]

model = DecisionTreeClassifier()
model.fit(X,y)

plt.figure(figsize=(10,10))
tree.plot_tree(model,class_names=y.unique(),feature_names=X.columns)
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from sklearn.model_selection import GridSearchCV
df = pd.read_csv(r"C:\Users\barat\Downloads\covtype (1).csv")
df=df.sample(50000)
X=df.drop(["Cover_Type"],axis=1)
y=df["Cover_Type"]
x_train, x_test, y_train, y_test = train_test_split(X,y,test_size=0.20)
model = DecisionTreeClassifier(max_depth=10,min_samples_split=15)
model.fit(x_train,y_train)
train_predictions=model.predict(x_train)
test_predictions=model.predict(x_test)
print(f"Train")
print(f"Train Accuracy: {accuracy_score(y_train, train_predictions)}")
print(f"Train Precision: {precision_score(y_train, train_predictions,average='micro')}")
print(f"Train Recall: {recall_score(y_train, train_predictions,average='micro')}")
print(f"Train F1 Score: {f1_score(y_train, train_predictions,average='micro')}")
print(f"Test")
print(f"Test Accuracy: {accuracy_score(y_test, test_predictions)}")
print(f"Test Precision: {precision_score(y_test, test_predictions,average='micro')}")
print(f"Test Recall: {recall_score(y_test, test_predictions,average='micro')}")
print(f"Test F1 Score: {f1_score(y_test, test_predictions,average='micro')}")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\barat\Downloads\Obesity prediction.csv")
df=pd.get_dummies(df,columns=["Gender","family_history","FAVC","CAEC","SMOKE","SCC","CALC","MTRANS"])
df.dropna(inplace=True)
X=df.drop(["Obesity"],axis=1)
y=df["Obesity"]
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=42)
models = [DecisionTreeClassifier(),RandomForestClassifier(),GradientBoostingClassifier()]
for model in models:
    model.fit(x_train,y_train)
    train_predictions=model.predict(x_train)
    test_predictions=model.predict(x_test)
    print(f"DecisionTreeClassifierResults")
    print(f"Train Accuracy for Decision Tree: {accuracy_score(y_train, train_predictions)}")
    print(f"Test Accuracy for Decision Tree: {accuracy_score(y_test, test_predictions)}")
    print(f"Train Precision for Decision Tree: {precision_score(y_train, train_predictions,average='micro')}")
    print(f"Test Precision for Decision Tree: {precision_score(y_test, test_predictions,average='micro')}")
    print(f"Train Recall for Decision Tree: {recall_score(y_train, train_predictions,average='micro')}")
    print(f"Test Recall for Decision Tree: {recall_score(y_test, test_predictions,average='micro')}")
    print(f"Train F1 Score for Decision Tree: {f1_score(y_train, train_predictions,average='micro')}")
    print(f"Test F1 Score for Decision Tree: {f1_score(y_test, test_predictions,average='micro')}")

    print(f"Results for Random Forest Classifier")
    print(f"Train Accuracy for Random Forest: {accuracy_score(y_train, train_predictions)}")
    print(f"Test Accuracy for Random Forest: {accuracy_score(y_test, test_predictions)}")
    print(f"Train Precision for Random Forest: {precision_score(y_train, train_predictions,average='micro')}")
    print(f"Test Precision for Random Forest: {precision_score(y_test, test_predictions,average='micro')}")
    print(f"Train Recall for Random Forest: {recall_score(y_train, train_predictions,average='micro')}")
    print(f"Test Recall for Random Forest: {recall_score(y_test, test_predictions,average='micro')}")
    print(f"Train F1 Score for Random Forest: {f1_score(y_train, train_predictions,average='micro')}")
    print(f"Test F1 Score for Random Forest: {f1_score(y_test, test_predictions,average='micro')}")

    print(f"Results for Gradient Boosting Classifier")
    print(f"Train Accuracy for Gradient Boosting: {accuracy_score(y_train, train_predictions)}")
    print(f"Test Accuracy for Gradient Boosting: {accuracy_score(y_test, test_predictions)}")
    print(f"Train Precision for Gradient Boosting: {precision_score(y_train, train_predictions,average='micro')}")
    print(f"Test Precision for Gradient Boosting: {precision_score(y_test, test_predictions,average='micro')}")
    print(f"Train Recall for Gradient Boosting: {recall_score(y_train, train_predictions,average='micro')}")
    print(f"Test Recall for Gradient Boosting: {recall_score(y_test, test_predictions,average='micro')}")
    print(f"Train F1 Score for Gradient Boosting: {f1_score(y_train, train_predictions,average='micro')}")
    print(f"Test F1 Score for Gradient Boosting: {f1_score(y_test, test_predictions,average='micro')}")
```
